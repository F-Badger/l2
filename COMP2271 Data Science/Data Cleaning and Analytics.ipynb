## Types of file

### Text file (.txt)
- reading and writing line by line
- files that humans can read "by eye"
- a standard text document that contains unformatted text

### Comma separated values (.csv)
- data fields are separated (delimited) by commas
- used to store tabular data, such as a spreadsheet or database
- CSV files can be imported to and exported from programs that store data in tables, such as Microsoft Excel

## Built-in functions for file operations (Python)

### Opening a file
```python
f = open(file, mode='r', buffering=-1, encoding=None, errors=None, newline=None, closefd=True, opener=None)
```
- `file` is the only required argument, and is the relative path to the file, including the extension
- `mode` specifies how the file is opened:
  - `r`: read. Show an error if the file does not exist. This is the default mode
  - `w`: write (overwrite). Create if file does not exist
  - `a` append (at the end) write. Create if file does not exist
  - `r+` both read and write. Show an error if file does not exist
  - `w+` both read and write. Create if file does not exist

### Closing a file
Files should be closed after the necessary operations involving them have been performed, in order to free system resources, ensure data is actually written to storage and to prevent file locks.

```python
f.close()
```

### Reading from a file
There are multiple methods for reading from a file:
```python
f.read() # read all the contents of the file as one string

f.readline() # single line reading (including a newline character (\n)). Return empty string when finished reading

f.readlines() # read all the lines of a file in a list
```

### Writing to a file
There are also multiple methods to write to a file:
```python
f.write(string) # writes the contents of string to the file, returning the number of characters written.

f.writelines(sequence) # writes a sequence of strings to the file. The sequence can be any iterable object producing strings, typically a list of strings. There is no return value.
```

## `with`

A `with` block can be used to open a file more safely. The file is automatically closed when the block finishes executing, whether an error occurs or not, so there is no need to call `f.close()` manually.

```python
with open("file.txt", "r") as f:
    data = f.read()
# file automatically closed here
```

## Extracting data from ZIP files

ZIP is an archive file format that supports lossless data compression. A ZIP file may contain one or more files or directories that may have been
compressed.

To work with ZIP files, the ZipFile class needs to be imported:
```python
from zipfile import ZipFile
```

Example usage of the class, starting with files `file1.txt` and `file2.txt`, is given below:

````python
# create a ZIP called example.zip containing files file1.txt and file2.txt
from zipfile import ZipFile
with ZipFile('example.zip', 'w') as myzip:
    myzip.write('file1.txt')
    myzip.write('file2.txt')
    myzip.close()
````

````python
# open the ZIP and perform operations on the contents
with ZipFile('example.zip', 'r') as myzip:
    # list of file names in the archive
    files = myzip.namelist()
    print("Files:", files)
    # the text lines of all files to be gathered into a single list
    textlines=[]
    for file in files:
        extrdata = myzip.read(file)
        print("Contents of", file + ":", extrdata)
        # the data in the archived file is stored in the byte format, it needs to be decoded back to utf-8
        origdata = extrdata.decode('utf-8')
        origlines = origdata.splitlines()
        textlines = textlines + origlines
    print ("All lines of all files:", textlines)
````
