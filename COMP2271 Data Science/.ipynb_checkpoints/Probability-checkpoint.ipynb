# Probability

## Combinatorics

The theory of counting is known as **combinatorics**. It deals with counting the number of ways that an event can occur. 

The main approach taken involves simplifying the solution by decomposing the problem into smaller parts.

### The product rule

If a procedure can be broken down into a sequence of two tasks, and there are $n_1$ ways to do the first task, and $n_2$ ways to do the second task, then the total number of ways to complete the procedure is $n_1 n_2$. 

This can be generalised to:
 -  If a sequence of tasks $T_1,T_2,\dots ,T_m$ can be done in $n_1,n_2, \dots, n_m$ ways respectively, and every task arrives after the occurrence of the previous task, then there are $n_1 \cdot n_2 \cdot \ldots \cdot n_m$ ways to perform the tasks.

### The sum rule

If a procedure requires only one of two independent tasks to be completed, and there are $n_1$ ways to do the first task, and $n_2$ ways to do the second task, then the total number of ways to complete the procedure is $n_1 + n_2$. 

This can be generalised to:
 -  If a sequence of tasks $T_1,T_2,\dots ,T_m$ can be done in $n_1,n_2, \dots, n_m$ ways respectively (with no tasks being performed simultaneously), then there are $n_1 + n_2 + \ldots + n_m$ ways to perform one of the tasks.

**The product rule is used when the sub-tasks of an event are dependent on one another (each of them is performed); the sum rule is used when they are independent (only one of them is performed)**

### Inclusion-exclusion principle

If two tasks can be performed at the same time, there is a possibility of overcounting, since some ways may be counted twice. To prevent this:
 - Add the number of ways to perform each task
 - Then subtract the number of ways to do both tasks

### Permutations

A permutation is an ordered arrangement of objects. An ordered arrangement of $r$ objects is called an $r$-permutation. 
 - ABC, ACB, BAC, BCA, CAB and CBA are all the unique 3-permutations of the set {A,B,C}

The number of $r$-permuations for a set of $n$ distinct elements is given by:
$$ P(n,r) = n \times (n-1) \times (n-2) \times \cdots \times (n-r+1) $$

This can be expressed more simply in factorial form:
$$ P(n,r) = \frac{n!}{(n-r)!} $$

This formula assumes all objects in the set are distinct. If the set contains $n$ objects, of which $n_1$ are of one type (indistinguishable from each other), $n_2$ are of a second type, $n_k$ are of a $k_{th}$ type, so that $n = n_1 + n_2 + \cdots + n_k$, then the number of distinct permutations is given by:
$$ P(n;n_1,n_2,\cdots,n_k) = \frac{n!}{{n_1}!{n_2}!\cdots{n_k}!} $$

### Combinations

A combination is an unordered arrangement of objects. An unordered arrangement of $r$ objects is called an $r$-combination. 
 - While $a,b$ and $b,a$ would be unique permutations, they represent the same combination.

The number of $r$-combinations for a set of $n$ distinct elements is given by:
$$ C(n,r) = {n \choose r} = \frac{n!}{(n-r)! \, r!} $$

## Set theory

A set is a collection of unordered objects.
The size, or cardinality, of a set $L$, $|L|$, is the number of elements in the set. 

For two sets $L$ and $R$, the basic set operations are:
 - Union: $L \cup R$ - the set of elements in $L$ or $R$ or both
 - Intersection: $L \cap R$ - the set of elements in both $L$ and $R$
 - Complement: $\overline{L}$ - all the elements not in $L$
 - Set Difference/Subtract: $L - R$ - elements in $L$ that are not in $R$
 - Subset: $L \subseteq S$ - every element of $L$ is also in $S$

## Theory of probability

An *experiment* is a procedure that yields one of a given set of outcomes, e.g. flipping a coin.
The *sample space* of an experiment is the set of possible outcomes, e.g. heads or tails.
An *event* is any subset of the sample space, e.g. heads. 

In a finite sample space $S$, where each outcome is equally likely, the probability of an event $E \subseteq S$ occuring is 
$$ P(E) = \frac{|E|}{|S|} $$

There are three axioms of probability:
 - $0 \leq P(E) \leq 1$ - the probability of any event occuring is always between 0 and 1
 - $P(S) = 1$ - the probability that some outcome in the sample space occurs is 1, i.e. the outcome is always in the sample space
 - For disjoint (mutually exclusive) events $E_1, E_2, \ldots, E_n$ (events with no intersection):
   - $P(E_1 \cup E_2 \cup \cdots \cup E_n) = \sum_{i=1}^{n} P(E_i)$

Other rules can be derived from these:
 - Complement rule: $P(\overline{E}) = 1 - P(E)$
 - Impossible event rule: $P(\emptyset) = 0$ - the probability of an event that can never occur is 0
 - General addition rule: $P(E_1 \cup E_2) = P(E_1) + P(E_2) - P(E_1 \cap E_2)$
 - Subset rule: If $E_1 \subseteq E_2$, then $P(E_1) \leq P(E_2)$

## Relationship of events

Events can happen concurrently, after each other or because of each other (or even more complicated).

Depending on the nature of the relationship, the probability of each event can be calculated based on other events.

### Conditional probability

Conditional probability is the probability that an event occurs given that another event has already occurred.

The probability of event A occuring given that event B has occured (the conditional probability) is given by:
$$ P(A|B) = \frac{P(A \cap B)}{P(B)} $$

Therefore, by multiplying by $P(B)$, the probability of A and B is:
$$ P(A \cap B) = P(B) \times P(A|B) $$
This is called the chain rule. 

### Dependence
Two events are independent if the occurrence of one event is not dependent on the occurrence of another.

Two events A and B are independent if and only if:
$$ P(A \cap B) = P(A) \times P(B) $$

Equivalently, if they are independent, the probability of A will not change given that B happens:
$$ P(A|B) = P(A) $$

### Bayes' Theorem
Bayes' theorem allows one conditional probability, $P(A | B)$, to be calculated from another, $P(B|A)$:
$$ P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$$

A general form of Bayes' theorem is as follows:
- suppose $E_1, E_2, \ldots, E_n$ are mutually exclusive events such that $\bigcup_{i=1}^{n} E_i = S$, where $S$ is the sample space
- then, if $A$ is any event, Bayes' Theorem states:
  $$P(E_k \mid A) = \frac{P(E_k) \times P(A \mid E_k)}{\sum_{j=1}^{n} (P(E_j) \times P(A \mid E_j))}$$

## Tree diagrams
A tree consists of a root, a number of branches, each leading to other branches, ending at leaf nodes. Branches are used for each possible choice.

To find the probability of a particular outcome (leaf), multiply the probabilities along the branches that lead to that outcome. 

## Random variables
A random variable is a function that assigns a number to each outcome in the sample space of a random experiment. It is usually denoted by a capital letter such as $X$ or $Y$.

A random variable $X$ is discrete if it has countably many values (e.g., $x_1, x_2, x_3, \ldots$).

For example, suppose a fair coin is tossed twice. The sample space will be {$HH, HT, TH, TT$}. A random variable $X$ can be created to count the number of heads that occur. $X$ will assign a numerical value to each outcome in the sample space:

| Outcome | $HH$ | $HT$ | $TH$ | $TT$ |
| ------- | --- | --- | --- | --- |
| $X$ | $2$ | $1$ | $1$ | $0$ |

## Probability distribution
The probability distribution of a random variable $X$ describes the probabilities associated with each possible outcome of $X$.

Let $X$ be a discrete random variable, which can assume values $x_1, x_2, \ldots, x_n$. Then, these values are associated with probabilities given by
$$ P(X = x_k) = f(x_k) , k = 1, 2, \ldots, n $$

This function $f$ is known as the probability mass function (PMF).

### Probability mass function
For a discrete random variable X with possible values of $x_1, x_2, \ldots$, the probability mass function is a function $f: x_1, x_2, \ldots → [0,1]$ satisfying three properties:
- $f(x_i) \geq 0$ - each probability is postive (a probability can't be negative)
- $\sum_{k=1}^\infty f(x_i) = 1$ - all the probabilities sum to 1
- $P(X=x_i) = f(x_i)$ - it associates a value of the random variable to the probability of that value

To compute a PMF ($P(X=x)$):
1. collect all possible outcomes where $X=x$
2. sum their probabilities
3. repeat for all $x$

Continuing with the example of tossing a fair coin twice:
| Outcome | $HH$ | $HT$ | $TH$ | $TT$ |
| ------- | --- | --- | --- | --- |
| $X$ | $2$ | $1$ | $1$ | $0$ |
| $P(X=x)$ | $\frac{1}{4}$ | $\frac{1}{2}$ | $\frac{1}{2}$ | $\frac{1}{4}$ |

### Cumulative distribution function
The cumulative distribution function (CDF) is a function $F : \mathbb{R} \to [0,1]$ that gives the probability that $X$ is less than or equal to a given value $x$. It is defined as follows:
$$ F(X) = P(X \le x) = \sum_{x_i \le x} P(X = x_i) = \sum_{x_i \le x} f(x_i), \quad \text{where } x \in \mathbb{R} $$

Note the use of $F$ for CDF and $f$ for PMF.

For discrete variables, PMF is used more than CDF.

All of the information contained in the PMF is also contained in the CDF. However, the CDF is properly defined for any random variable and any real number, and not only for the values $X$ can take, unlike the PMF which is restricted to the values $X$ can take.

## Mean and variance
Mean and variance are often used to summarise a probability distribution for a random variable $X$.

The mean (or expected value) is the centre/middle of the probability distribution:
$$ \mu = E(X) = \sum_{x} x f(x) $$

Variance is a measure of the dispersion/variability in the distribution:
$$
\sigma^2 = {V}(X)
= {E}\!\left[(X-\mu)^2\right]
= \sum_x (x-\mu)^2 f(x)
= \sum_x x^2 f(x) - \mu^2
$$

Standard deviation is the square root of variance:
$$ \sigma = \sqrt{\sigma^2} $$

## Functions of random variables
Functions of random variables are also random variables:
- let $X$ be a random variable and let $h: \mathbb{R} → \mathbb{R}$ be an arbitrary function, then $Y = h(X)$ is also a random variable

Let $X$ be a random variable with PMF $f$ and $h: \mathbb{R} → \mathbb{R}$ be an arbitrary function, then $Y = h(X)$ is also a random variable. If $X$ is discrete and takes values $\{x_1, x_2, \ldots \}$ then:
$$ E[Y] = E[h(X)] = \sum_i h(x_i)f(x_i) $$

These laws mean that a new random variable can be formed by performing arithmetic operations on a random variable, and the expectation (mean) and variance of the new variable can be computed from the originals. 

Some properties that can be derived from this information include:
- $E(aX + b) = aE(X) + b$
- $E[h(X)] = \sum_x h(x) f(x)$
- $V(aX + b) = a^2 V(X)$
- $V(X) = E[X^2] - (E[X])^2$

$E[X^2]$ is known as the second moment, and $E[X^2]$ as the first moment. The variance is also called the central second moment. 

## Independence of random variables
Two random variables $X$ and $Y$ are independent if for any sets $A$ and $B$:
$$
P(X \in A \cap Y \in B) \equiv P(X \in A, Y \in B) = P(X \in A)\, P(Y \in B)
$$

For independent random variables $X_1, X_2, \ldots, X_n$:
$$
V(X_1 + X_2 + \cdots + X_n) = V(X_1) + V(X_2) + \cdots + V(X_n)
$$

For any random variables (regardless of dependence!):
$$
E(X_1 + X_2 + \cdots + X_n) = E(X_1) + E(X_2) + \cdots + E(X_n)
$$

## Special discrete distributions
It is probably not needed to memorise any of the specific 'shortcut' formulae for any of these, since the values can be calculated using the general formulas specified before.

### Discrete uniform distribution
A random variable $X$ has a discrete uniform distribution if each of the $n$ values in its range $x_1, x_2, \ldots, x_n$ has equal probability.

Suppose the random variable X has uniform distribution. The PMF of X is given by ($X \sim \text{Uni}(n)$):
$$ P(X = k) = f(k) = \frac{1}{n} $$

Suppose $X$ is a discrete uniform random variable on the consecutive integers $a,a+1, a+2, \cdots, b$ for 𝑎 ≤ 𝑏. Then:
$$
\mu = \mathbb{E}[X] = \frac{a+b}{2}
$$
$$
\sigma^2 = \text{Var}(X) = \frac{(b-a+1)^2 - 1}{12}
$$

### Bernoulli distribution
In a random experiment, if each outcome consists of only two possibilities in each trial (such as in a toss of a coin), then such a random experiment is called a Bernoulli trial. Often one outcome is called 'success' and the other 'failure'.

In a Bernoulli trial, let the possible events in each outcome be denoted by $A$ and $B$, and let $P(A) = p$. Then:
- $A \cap B = \emptyset$
- $A \cup B = S$
- $P(B) = P(\bar{A}) = 1 - P(A) = 1 - p = q$
- 
Let $X$ be a random variable taking only two possible values $\{0,1\}$. Let $p = P(X=1)$. Then $X$ is said to be a Bernoulli random variable with parameter $p$:
$$ X \sim \text{Ber}(p) $$
$$ E[X] = p $$
$$ \text{Var}(X) = p(1-p) = pq $$

### Discrete binomial distribution
Consider a random experiment consisting of $n \in \mathbb{N}$ independent Bernoulli trials, each with probability of success $p$ $(0 \le p \le 1)$, constant across trials.

Let $X$ denote the number of successes in these $n$ trials. Then $X$ is a Binomial random variable:
$$
X \sim \text{Bin}(n, p)
$$
$$
P(X = k) = f(k) = \binom{n}{k} p^k (1-p)^{n-k}, \quad k = 0, 1, \dots, n
$$
$$
\mathbb{E}[X] = np
$$
$$
\text{Var}(X) = np(1-p)
$$

### Discrete geometric distribution
Consider a series of independent Bernoulli trials with constant probability of success $p$. 
Let the random variable $X$ denote the number of trials until the first success. 

Then $X$ is called a geometric random variable with parameter $p$:
$$
X \sim \text{Geom}(p)
$$
$$
P(X = k) = f(k) = p (1-p)^{k-1}, \quad k = 1, 2, \dots
$$
$$
\mathbb{E}[X] = \frac{1}{p}
$$
$$
\text{Var}(X) = \frac{1-p}{p^2}
$$

The geometric distribution is the only discrete distribution satisfying the memoryless property. This means that the probability of success in the future does not depend on the past:
$$ P(X \gt m+n ∣ X \gt m) = P (X \gt n) $$

### Negative binomial distribution
The negative binomial distribution generalises the geometric distribution.  
Instead of counting the number of trials until the first success, count the number of trials until the $r$-th success occurs, where $r \in \mathbb{N}$.  

Consider a sequence of independent Bernoulli trials, each with the same probability of success $p$.  
Let the random variable $X$ denote the total number of trials needed to achieve $r$ successes.

Then $X$ is called a negative binomial random variable with parameters $r$ and $p$:
$$
X \sim \text{NegBin}(r, p)
$$
$$
P(X = k) = f(k) =
\begin{cases}
0, & k < r \\
\binom{k-1}{r-1} p^r (1-p)^{k-r}, & k \ge r
\end{cases}, \quad k \in \mathbb{N}
$$
$$
\mathbb{E}[X] = \frac{r}{p}
$$
$$
\text{Var}(X) = \frac{r(1-p)}{p^2}
$$
