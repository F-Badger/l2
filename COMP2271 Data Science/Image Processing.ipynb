# Image representation
- Images are 2D signals.
- Signals can be digitised as a structured set of numbers.
- Therefore, images can be represented by 2D arrays or matrices.

# Image transformation
If an image is a matrix, then the image can be transformed by manipulating the corresponding matrix. 

## Types of transformation
There are three types of image transformation:
- a point transform involving only a single pixel at a time
- a local transform involving the local image neighbourhood (a pixel and those immediately next to it)
- a global transform involving the whole image
