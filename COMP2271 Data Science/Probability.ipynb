# Probability

## Combinatorics

The theory of counting is known as **combinatorics**. It deals with counting the number of ways that an event can occur. 

The main approach taken involves simplifying the solution by decomposing the problem into smaller parts.

### The product rule

If a procedure can be broken down into a sequence of two tasks, and there are $n_1$ ways to do the first task, and $n_2$ ways to do the second task, then the total number of ways to complete the procedure is $n_1 n_2$. 

This can be generalised to:

 -  If a sequence of tasks $T_1,T_2,\dots ,T_m$ can be done in $n_1,n_2, \dots, n_m$ ways respectively, and every task arrives after the occurrence of the previous task, then there are $n_1 \cdot n_2 \cdot \ldots \cdot n_m$ ways to perform the tasks.

### The sum rule

If a procedure requires only one of two independent tasks to be completed, and there are $n_1$ ways to do the first task, and $n_2$ ways to do the second task, then the total number of ways to complete the procedure is $n_1 + n_2$. 

This can be generalised to:

 -  If a sequence of tasks $T_1,T_2,\dots ,T_m$ can be done in $n_1,n_2, \dots, n_m$ ways respectively (with no tasks being performed simultaneously), then there are $n_1 + n_2 + \ldots + n_m$ ways to perform one of the tasks.

**The product rule is used when the sub-tasks of an event are dependent on one another (each of them is performed); the sum rule is used when they are independent (only one of them is performed)**

### Inclusion-exclusion principle

If two tasks can be performed at the same time, there is a possibility of overcounting, since some ways may be counted twice. To prevent this:

 - Add the number of ways to perform each task

 - Then subtract the number of ways to do both tasks

### Permutations

A permutation is an ordered arrangement of objects. An ordered arrangement of $r$ objects is called an $r$-permutation. 
 - ABC, ACB, BAC, BCA, CAB and CBA are all the unique 3-permutations of the set {A,B,C}

The number of $r$-permuations for a set of $n$ distinct elements is given by:
$$ P(n,r) = n \times (n-1) \times (n-2) \times \cdots \times (n-r+1) $$

This can be expressed more simply in factorial form:
$$ P(n,r) = \frac{n!}{(n-r)!} $$

This formula assumes all objects in the set are distinct. If the set contains $n$ objects, of which $n_1$ are of one type (indistinguishable from each other), $n_2$ are of a second type, $n_k$ are of a $k_{th}$ type, so that $n = n_1 + n_2 + \cdots + n_k$, then the number of distinct permutations is given by:
$$ P(n;n_1,n_2,\cdots,n_k) = \frac{n!}{{n_1}!{n_2}!\cdots{n_k}!} $$

### Combinations

A combination is an unordered arrangement of objects. An unordered arrangement of $r$ objects is called an $r$-combination. 

 - While $a,b$ and $b,a$ would be unique permutations, they represent the same combination.

The number of $r$-combinations for a set of $n$ distinct elements is given by:
$$ C(n,r) = {n \choose r} = \frac{n!}{(n-r)! \, r!} $$

## Set theory

A set is a collection of unordered objects.
The size, or cardinality, of a set $L$, $|L|$, is the number of elements in the set. 

For two sets $L$ and $R$, the basic set operations are:

 - Union: $L \cup R$ - the set of elements in $L$ or $R$ or both
   
 - Intersection: $L \cap R$ - the set of elements in both $L$ and $R$

 - Complement: $\overline{L}$ - all the elements not in $L$
   
 - Set Difference/Subtract: $L - R$ - elements in $L$ that are not in $R$

 - Subset: $L \subseteq S$ - every element of $L$ is also in $S$

## Theory of probability

An *experiment* is a procedure that yields one of a given set of outcomes, e.g. flipping a coin.
The *sample space* of an experiment is the set of possible outcomes, e.g. heads or tails.
An *event* is any subset of the sample space, e.g. heads. 

In a finite sample space $S$, where each outcome is equally likely, the probability of an event $E \subseteq S$ occuring is 
$$ P(E) = \frac{|E|}{|S|} $$

There are three axioms of probability:

 - $0 \leq P(E) \leq 1$ - the probability of any event occuring is always between 0 and 1
   
 - $P(S) = 1$ - the probability that some outcome in the sample space occurs is 1, i.e. the outcome is always in the sample space

   
 - For disjoint (mutually exclusive) events $E_1, E_2, \ldots, E_n$ (events with no intersection):


   - $P(E_1 \cup E_2 \cup \cdots \cup E_n) = \sum_{i=1}^{n} P(E_i)$

Other rules can be derived from these:

 - Complement rule: $P(\overline{E}) = 1 - P(E)$
 - Impossible event rule: $P(\emptyset) = 0$ - the probability of an event that can never occur is 0
 - General addition rule: $P(E_1 \cup E_2) = P(E_1) + P(E_2) - P(E_1 \cap E_2)$
 - Subset rule: If $E_1 \subseteq E_2$, then $P(E_1) \leq P(E_2)$

## Relationship of events

Events can happen concurrently, after each other or because of each other (or even more complicated).

Depending on the nature of the relationship, the probability of each event can be calculated based on other events.

### Conditional probability

Conditional probability is the probability that an event occurs given that another event has already occurred.

The probability of event A occuring given that event B has occured (the conditional probability) is given by:
$$ P(A|B) = \frac{P(A \cap B)}{P(B)} $$

Therefore, by multiplying by $P(B)$, the probability of A and B is:
$$ P(A \cap B) = P(B) \times P(A|B) $$
This is called the chain rule. 

### Dependence
Two events are independent if the occurrence of one event is not dependent on the occurrence of another.

Two events A and B are independent if and only if:
$$ P(A \cap B) = P(A) \times P(B) $$

Equivalently, if they are independent, the probability of A will not change given that B happens:
$$ P(A|B) = P(A) $$

### Bayes' Theorem
Bayes' theorem allows one conditional probability, $P(A | B)$, to be calculated from another, $P(B|A)$:
$$ P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$$

A general form of Bayes' theorem is as follows:

- suppose $E_1, E_2, \ldots, E_n$ are mutually exclusive events such that $\bigcup_{i=1}^{n} E_i = S$, where $S$ is the sample space
- then, if $A$ is any event, Bayes' Theorem states:
  $$P(E_k \mid A) = \frac{P(E_k) \times P(A \mid E_k)}{\sum_{j=1}^{n} (P(E_j) \times P(A \mid E_j))}$$

## Random variables
A random variable is a function that assigns a number to each outcome in the sample space of a random experiment. It is usually denoted by a capital letter such as $X$ or $Y$.

A random variable $X$ is discrete if it has countably many values (e.g., $x_1, x_2, x_3, \ldots$).

For example, suppose a fair coin is tossed twice. The sample space will be {$HH, HT, TH, TT$}. A random variable $X$ can be created to count the number of heads that occur. $X$ will assign a numerical value to each outcome in the sample space:

- HH: 2  
- HT: 1  
- TH: 1  
- TT: 0

## Probability distribution
The probability distribution of a random variable $X$ describes the probabilities associated with each possible outcome of $X$.

Let $X$ be a discrete random variable, which can assume values $x_1, x_2, \ldots, x_n$. Then, these values are associated with probabilities given by
$$ P(X = x_k) = f(x_k) , k = 1, 2, \ldots, n $$

This function $f$ is known as the probability mass function (PMF).

### Probability mass function
For a discrete random variable X with possible values of $x_1, x_2, \ldots$, the probability mass function is a function $f: x_1, x_2, \ldots → [0,1]$ satisfying three properties:

- $f(x_i) \geq 0$ - each probability is positive (a probability can't be negative)
- $\sum_{k=1}^\infty f(x_i) = 1$ - all the probabilities sum to 1
- $P(X=x_i) = f(x_i)$ - it associates a value of the random variable to the probability of that value

To compute a PMF ($P(X=x)$):

1. collect all possible outcomes where $X=x$
2. sum their probabilities
3. repeat for all $x$

Continuing with the example of tossing a fair coin twice:

- HH: X = 2, P(X=2) = 1/4
- HT: X = 1, P(X=1) = 1/2
- TH: X = 1, P(X=1) = 1/2
- TT: X = 0, P(X=0) = 1/4

### Cumulative distribution function
The cumulative distribution function (CDF) is a function $F : \mathbb{R} \to [0,1]$ that gives the probability that $X$ is less than or equal to a given value $x$. It is defined as follows:
$$ F(X) = P(X \le x) = \sum_{x_i \le x} P(X = x_i) = \sum_{x_i \le x} f(x_i), \quad \text{where } x \in \mathbb{R} $$

Note the use of $F$ for CDF and $f$ for PMF.

For discrete variables, PMF is used more than CDF.

All of the information contained in the PMF is also contained in the CDF. However, the CDF is properly defined for any random variable and any real number, and not only for the values $X$ can take, unlike the PMF which is restricted to the values $X$ can take.

## Mean and variance
Mean and variance are often used to summarise a probability distribution for a random variable $X$.

The mean (or expected value) is the centre/middle of the probability distribution:
$$ \mu = E(X) = \sum_{x} x f(x) $$

Variance is a measure of the dispersion/variability in the distribution:
$$
\sigma^2 = {V}(X)
= {E}\!\left[(X-\mu)^2\right]
= \sum_x (x-\mu)^2 f(x)
= \sum_x x^2 f(x) - \mu^2
$$

Standard deviation is the square root of variance:
$$ \sigma = \sqrt{\sigma^2} $$

## Functions of random variables
Functions of random variables are also random variables:

- let $X$ be a random variable and let $h: \mathbb{R} → \mathbb{R}$ be an arbitrary function, then $Y = h(X)$ is also a random variable

Let $X$ be a random variable with PMF $f$ and $h: \mathbb{R} → \mathbb{R}$ be an arbitrary function, then $Y = h(X)$ is also a random variable. If $X$ is discrete and takes values $\{x_1, x_2, \ldots \}$ then:
$$ E[Y] = E[h(X)] = \sum_i h(x_i)f(x_i) $$

These laws mean that a new random variable can be formed by performing arithmetic operations on a random variable, and the expectation (mean) and variance of the new variable can be computed from the originals. 

Some properties that can be derived from this information include:
- $E(aX + b) = aE(X) + b$
- $E[h(X)] = \sum_x h(x) f(x)$
- $V(aX + b) = a^2 V(X)$
- $V(X) = E[X^2] - (E[X])^2$

$E[X^2]$ is known as the second moment, and $E[X]$ as the first moment. The variance is also called the central second moment. 

## Independence of random variables
Two random variables $X$ and $Y$ are independent if for any sets $A$ and $B$:
$$
P(X \in A \cap Y \in B) \equiv P(X \in A, Y \in B) = P(X \in A)\, P(Y \in B)
$$

For independent random variables $X_1, X_2, \ldots, X_n$:
$$
V(X_1 + X_2 + \cdots + X_n) = V(X_1) + V(X_2) + \cdots + V(X_n)
$$

For any random variables (regardless of dependence!):
$$
E(X_1 + X_2 + \cdots + X_n) = E(X_1) + E(X_2) + \cdots + E(X_n)
$$

## Special discrete distributions
The formulae for the PMF, expectation and variance of all the specific distributions below, apart from discrete uniform distribution, are given in the exam on the formulae sheet. It is therefore not needed to memorise them.

### Discrete uniform distribution
A random variable $X$ has a discrete uniform distribution if each of the $n$ values in its range $x_1, x_2, \ldots, x_n$ has equal probability.

Suppose the random variable X has uniform distribution. The PMF of X is given by ($X \sim \text{Uni}(n)$):
$$ P(X = k) = f(k) = \frac{1}{n} $$

Suppose $X$ is a discrete uniform random variable on the consecutive integers $a,a+1, a+2, \cdots, b$ for $a \le b$. Then:
$$
\mu = \mathbb{E}[X] = \frac{a+b}{2}
$$
$$
\sigma^2 = \text{Var}(X) = \frac{(b-a+1)^2 - 1}{12}
$$

### Bernoulli distribution
In a random experiment, if each outcome consists of only two possibilities in each trial (such as in a toss of a coin), then such a random experiment is called a Bernoulli trial. Often one outcome is called 'success' and the other 'failure'.

In a Bernoulli trial, let the possible events in each outcome be denoted by $A$ and $B$, and let $P(A) = p$. Then:

- $A \cap B = \emptyset$
- $A \cup B = S$
- $P(B) = P(\bar{A}) = 1 - P(A) = 1 - p = q$

Let $X$ be a random variable taking only two possible values $\{0,1\}$. Let $p = P(X=1)$. Then $X$ is said to be a Bernoulli random variable with parameter $p$:
$$ X \sim \text{Ber}(p) $$
$$ E[X] = p $$
$$ \text{Var}(X) = p(1-p) = pq $$

### Discrete binomial distribution
Consider a random experiment consisting of $n \in \mathbb{N}$ independent Bernoulli trials, each with probability of success $p$ $(0 \le p \le 1)$, constant across trials.

Let $X$ denote the number of successes in these $n$ trials. Then $X$ is a Binomial random variable:
$$
X \sim \text{Bin}(n, p)
$$
$$
P(X = k) = f(k) = \binom{n}{k} p^k (1-p)^{n-k}, \quad k = 0, 1, \dots, n
$$
$$
\mathbb{E}[X] = np
$$
$$
\text{Var}(X) = np(1-p)
$$

### Discrete geometric distribution
Consider a series of independent Bernoulli trials with constant probability of success $p$. 
Let the random variable $X$ denote the number of trials until the first success. 

Then $X$ is called a geometric random variable with parameter $p$:
$$
X \sim \text{Geom}(p)
$$
$$
P(X = k) = f(k) = p (1-p)^{k-1}, \quad k = 1, 2, \dots
$$
$$
\mathbb{E}[X] = \frac{1}{p}
$$
$$
\text{Var}(X) = \frac{1-p}{p^2}
$$

The geometric distribution is the only discrete distribution satisfying the memoryless property. This means that the probability of success in the future does not depend on the past:
$$ P(X \gt m+n ∣ X \gt m) = P (X \gt n) $$

### Negative binomial distribution
The negative binomial distribution generalises the geometric distribution.  
Instead of counting the number of trials until the first success, count the number of trials until the $r$-th success occurs, where $r \in \mathbb{N}$.  

Consider a sequence of independent Bernoulli trials, each with the same probability of success $p$.  
Let the random variable $X$ denote the total number of trials needed to achieve $r$ successes.

Then $X$ is called a negative binomial random variable with parameters $r$ and $p$:
$$
X \sim \text{NegBin}(r, p)
$$
$$
P(X = k) = f(k) =
\begin{cases}
0, & k < r \\
\binom{k-1}{r-1} p^r (1-p)^{k-r}, & k \ge r
\end{cases}, \quad k \in \mathbb{N}
$$
$$
\mathbb{E}[X] = \frac{r}{p}
$$
$$
\text{Var}(X) = \frac{r(1-p)}{p^2}
$$

### Poisson distribution

The Poisson distribution models the number of events occurring in a fixed interval of time or space.  
Instead of counting trials until success, it counts how many times an event occurs when events happen independently at a constant average rate $\lambda > 0$.

Consider a process where events occur randomly and independently, with an average rate of $\lambda$ events per interval.  
Let the random variable $X$ denote the number of events in a given interval.

Then $X$ is called a Poisson random variable with parameter $\lambda$:
$$
X \sim \text{Poisson}(\lambda)
$$
$$
P(X = k) = f(k) =
\begin{cases}
\frac{\lambda^k e^{-\lambda}}{k!}, & k \ge 0 \\
0, & k < 0
\end{cases}, \quad k \in \mathbb{N}_0
$$
$$
\mathbb{E}[X] = \lambda
$$
$$
\text{Var}(X) = \lambda
$$

**Key properties:**

- Events occur independently  
- The rate $\lambda$ is constant over the interval  
- The probability of more than one event in a very small interval is negligible
- The interarrival times of a Poisson process have a (continuous) exponential distribution with mean $\frac{1}{\lambda}$.

Take $Y ∼ \text{Bin} (n, p)$ with large $n$, small $p$, such that $np ≃ \lambda$. Then $Y$ is approximately $\text{Poisson}(\lambda)$ distributed

- this means the Poisson distribution can be used to approximate the binomial distribution

## Continuous distributions

A random variable $X$ is continuous if it can take infinitely many values within a range (including decimals). 

Since it can have infinitely many values, the probability of $X$ taking any specific value is 0. Therefore, rather than considering the probability of $X$ taking a particular value, we consider the probability of $X$ lying within a certain interval. 

### Probability density function 

A probability density function (PDF) describes how a continuous random variable is distributed. 

The PDF of X is denoted by $f_X$ or simply by $f$.

The probability that the random variable lies within an interval is given by the area under the curve of the PDF over that interval (and thus can be obtained using integration).

![Porbability Density Function example](probability_density_function.png)

For a continuous random variable $X$, a probability density function is a function $f(x)$ such that:

1. $f(x) \ge 0$, for all $x \in \mathbb{R}$

2. $\int_{-\infty}^{+\infty} f(x)\,dx = 1$

3. $P(a \le X \le b) = \int_a^b f(x)\,dx, \quad \text{for any } a, b \in \mathbb{R} \text{ such that } a \le b$

Note that since the probability of $X$ taking any specific value is 0: 
$$P(a \le X \le b) = P(a < X \le b) = P(a < X < b) = P(a \le X < b)$$

### PDF and PMF

The PMF for discrete variables and the PDF for continuous variables play a similar role: they both describe how probabilities are distributed, but one uses sums and the other uses integrals.

![PDF and PMF side-by-side](pdf_pmf.png)

### Cumulative distribution function

Just as for discrete variables, the cumulative distribution function (CDF) for a continuous variable gives the probability that a random variable takes a value less than or equal to a certain number.

CDF is a function $F : \mathbb{R} \to [0,1]$ that gives the probability that $X$ is less than or equal to a given value $x$. It is defined as follows:
$$ F(X) = P(X \le x) = \int_{- \infty}^x f(t) dt $$

As in the discrete case, a continuous CDF has the following properties:

- Non-decreasing: if $x \le y$, then $F(x) \le F(y)$
- $0 \le F(x) \le 1$, $\lim_{x \to -\infty} F(x) = 0$ and $\lim_{x \to +\infty} F(x) = 1$
- for $a \le b$, $P(a \le X \le b) = F(b) - F(a)$

For a continuous variable, the derivative of the CDF is the PDF.

### Mean and variance

As with discrete variables, mean and variance are used to summarise the distribution:
$$ \mu = E(X) = \int_{- \infty}^{+ \infty} xf(x) dx$$

$$
\sigma^2 = E[(X - \mu)^2] = \int_{-\infty}^{+\infty} (x - \mu)^2 f(x)\,dx \quad \text{, or equivalently}
$$

$$
\sigma^2 = E[X^2] - \mu^2 = \int_{- \infty}^{+ \infty} x^2 f(x) dx - \mu ^ 2
$$

### Functions of random continuous variables

Let $X$ be a random variable and let $h: \mathbb{R} → \mathbb{R}$ be an arbitrary function, then $Y = h(X)$ is also a random variable. 

The properties are the same for continous variables as for [discrete variables](#Functions-of-random-variables), but the mean is calculated using an integral rather than a sum:
$$ E[Y] = E[h(X)] = \int_{-\infty}^{+\infty} h(x) f(x)\,dx $$

These proprties include:

- $E(aX + b) = aE(X) + b$
- $V(aX + b) = a^2 V(X)$
- $V(X) = E[X^2] - (E[X])^2$

### Specific continuous distributions

The formulae for the PDF, expectation and variance of all the specific distributions below, are given in the exam on the formulae sheet. It is therefore not needed to memorise them.

#### Continuous uniform distribution

A continuous uniform distribution models a random variable that is equally likely to take any value within a given interval $[a, b]$.

Let $X$ be a continuous random variable with density function:

$$
f(x) =
\begin{cases}
\frac{1}{b - a}, & a \le x \le b \\
0, & \text{otherwise}
\end{cases}
$$
Then, $X$ is said to be a continuous uniform random variable with range $[a,b]$:

$$
\mathbb{E}[X] = \frac{a+b}{2}
$$
$$
\mathbb{E}[X^2] = \frac{a^2 + ab + b^2}{3}
$$
$$
\text{Var}(X) = \frac{(b-a)^2}{12}
$$

#### Exponential distribution

A continuous random variable $X$ is called an exponential random variable with parameter $\lambda > 0$, (denoted by $X \sim \text{Exp}(\lambda)$) when its probability density function is:

$$
f(x) =
\begin{cases}
\lambda e^{-\lambda x}, & x \ge 0 \\
0, & x < 0
\end{cases}
$$

Here, $\lambda$ is called the rate parameter, and is the average number of events occuring in one period of unit time. This $\lambda$ is the same as in the [Poisson distribution](#Poisson-distribution); a variable $X \sim \text{Exp}(\lambda)$ models the time between events in a Poisson process with rate $\lambda$.

For the exponential distribution:
$$ \mathbb{E}[X] = \frac{1}{\lambda} $$
$$ \mathbb{E}[X^2] = \frac{2}{\lambda^2} $$
$$ \text{Var}(X) = \frac{1}{\lambda^2} $$

The exponential distribution is the only continuous distribution that satisfies the memoryless property:

- For any $t_1, t_2 \geq 0$:
  $$ P(X > t_1 + t_2 \; | \; X > t_1) = P(X > t_2) $$
- If $X$ denotes the time until a certain event happens, if we have already waited $t_1$ for the event, then the probability that we must wait $t_2$
more time does not depend on the value of $t_1$

## The normal distribution

The normal distribution (also called the Gaussian distribution) is a continuous distribution with significant importance, describing data that clusters around a central value with a symmetric 'bell-shaped' curve. 

The sum (or average) of many independent random variables tends to look normal (like a bell curve), regardless of the original distributions.

A random variable $X$ is said to be normally distributed if:
$$
X \sim \mathcal{N}(\mu, \sigma^2)
$$

where:

- $\mu \in (-\infty, \infty)$ is the mean (centre)
- $\sigma^2 > 0$ is the variance (spread)

and its probability density function (PDF) is:
$$
f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \; e^{-\frac{(x - \mu)^2}{2\sigma^2}}
$$

For a normally distributed variable $X$:
$$
E(X) = \mu
$$
$$
Var(X) = \sigma ^ 2
$$

It forms a bell shaped curve which is symmetric about $\mu$:

- $\mu$ controls the position of the curve (shifts it left/right)
- $\sigma$ controls the spread of the curve (larger $\sigma \to$ wider and flatter curve)
- the total area under the curve is 1

![Normal distribution bell curves](bell_curves.png)

### The empirical rule

For a normal distribution:
- About 68% of values (area) lie within $\mu \pm \sigma$
- About 95% lie within $\mu \pm 2\sigma$
- About 99.7% lie within $\mu \pm 3\sigma$

These results are known as the empirical rule. 

![Amount of data with 1,2 and 3 standard deviations](normal_area.png)

### Standard normal distribution
The best way to find a probability in a normal distribution is to convert it to the standard normal distribution. 

A random variable $Z$ is said to have a standard normal distribution if:
$$
Z \sim \mathcal{N}(0, 1)
$$

where:

- the mean ($\mu$) is $0$
- the variance ($\sigma^2$) is $1$

and its probability density function (PDF) is:
$$
\phi(z) = \frac{1}{\sqrt{2\pi}} \; e^{-\frac{z^2}{2}}
$$

For a standard normally distributed variable $Z$:
$$
E(Z) = 0
$$
$$
Var(Z) = 1
$$

All normal distributions $X$ can be converted into the standard normal distribution $Z$ by subtracting the mean and dividing by the standard
deviation:
$$
Z = \frac{X - \mu}{\sigma}
$$

- this is called Z-score:
  - it gives the number of standard deviations the value of $X$ is above or below the mean

The standard normal distribution makes it much easier to compare values from normal distributions using different scales, since they can be converted to use the same scale. 

#### Z-table
Integrals for the standard normal distribution have been computed and placed in a table called the Z-table, meaning that it is not necessary to compute integrals in order to determine probabilities after converting to the standard normal distribution; the probabilities can be looked up in the table. 

The Z-table provided in the exam gives the area to the to the left of each Z-score.

![Area to left of Z-score, giving P(Z < z)](z_curve.png)

To read the Z-table:

- Split the z-score after the first decimal point
  - the first digit and first decimal give the row
  - the second decimal gives the column
- Identify the value in this row and column
- This value gives the total area to the left of the z-score
  - this is $P(Z < z)$
  - $P(Z > z) = 1 - P(Z < z)$
  - $P(a < Z < b) = P(Z < b) - P(Z < a)$

### Normal approximation to the binomial distribution

If $X \sim \text{Bin}(n,p)$ then for large $n$, the distribution of $X$ looks approximately normal:
$$
Y \sim \mathcal{N}(np, np(1-p)) \quad \text{with } X \approx Y
$$


This approximation works well when:

- $np > 5$
- $n(1-p) > 5$

This ensures the binomial distribution is not too skewed.

Since the binomial distribution is discrete, while the normal distribution is continuous, a continuity correction should be applied when using normal approximation:
$$ P(X \leq k) \approx P(Y \leq k + 0.5)$$
$$ P(X < k) \approx P(Y \leq k - 0.5)$$

### Normal approximation to Poisson distribution

If $X \sim \text{Poisson}(\lambda)$ then for large $\lambda$, the distribution of $X$ looks approximately normal:
$$
Y \sim \mathcal{N}(\lambda, \lambda) \quad \text{with } X \approx Y
$$

A continuity correction must also be applied in the same way as with approximating a binomial distribution. 

## Multiple random variables

Multiple random variables are used to model and calculate the probability of events that depend on or occur together, capturing relationships between them rather than treating them in isolation.

Joint distributions enable reasoning about the relationship between multiple variables:

- whether or not they are dependent or independent
- conditional probabilities

Distributions with two variables are called bivariate distributions. Distributions with more than two variables are called multivariate distributions.

### Discrete bivariate joint distribution

The probability mass function (PMF) of a single discrete random variable $X$ specifies how much probability mass is placed on each possible $X$ value.

- The joint PMF of two discrete random variables $X$ and $Y$ describes how much probability mass is placed on each possible pair of values $(x, y)$

If $X$ and $Y$ are discrete random variables, the function given by
$$f(x,y) = P(X=x, Y=y)$$ 
for each pair of values $(x,y)$ is called the joint probability distribution (joint probability mass function) of $X$ and $Y$. It has the following properties:

- $f(x,y) \geq 0$
- $\sum_{x} \sum_{y} f(x,y) = 1$
- $f(x, y) = P(X = x, Y = y) ≡ P( \{X = x\} \cap \{Y = y\} )$

The joint CDF is the defined by:
$$
F(x, y) = P(X \leq x, Y \leq y) = \sum_{s \leq x} \sum_{t \leq y} f(s,t) \ , \ - \infty < x,y < \infty
$$

### Discrete multivariate joint distributions

The concept of a joint distribution extends to more than two random variables.

- If $X_1, X_2,\ldots, X_n$ are all discrete random variables, the joint PMF is:
$$
f(x_1, x_2, \ldots, x_n) = P(X_1 = x_1, X_2 = x_2, \ldots, X_n = x_n)
$$

### Discrete marginal probability distributions

A marginal probability distribution is the distribution of a single random variable obtained from a joint distribution by removing the other variable(s).

For a discrete bivariate joint distribution, this is done by summation:

- Let $X$ and $Y$ be two discrete random variables with joint probability mass function $f_{XY}(x, y)$. Then, the corresponding marginal probability mass functions are given by:
$$
f_X(x) = P(X = x) = \sum_y P(X = x, Y = y) = \sum_y f_{XY}(x,y)
$$
$$
f_Y(y) = P(Y = y) = \sum_x P(X = x, Y = y) = \sum_x f_{XY}(x,y)
$$ 



### Discrete conditional distributions

A conditional distribution describes the probability distribution of one random variable given that another random variable has taken a specific value.

Let $X$ and $Y$ be two discrete random variables with joint probability mass function $f_{XY}(x, y)$.

Assume $f_X(x) > 0$ (meaning $X = x$ is possible), then:

$$
f_{Y|X=x}(y) = P(Y = y \mid X = x) = \frac{P(X = x, Y = y)}{P(X = x)} = \frac{f_{XY}(x, y)}{f_X(x)}
$$

Similarly, if $f_Y(y) > 0$:

$$
f_{X|Y=y}(x) = P(X = x \mid Y = y) = \frac{P(X = x, Y = y)}{P(Y = y)} = \frac{f_{XY}(x, y)}{f_Y(y)}
$$

The conditional distribution is a valid PMF, so:
1. $\sum_{y \in Y} f_{Y|X=x}(y) = 1$

2. For any set $B$:
$$
P(Y \in B \mid X = x) = \sum_{y \in B} f_{Y|X=x}(y)
$$



### Continuous bivariate joint distributions

The probability density function (PDF) of a single continuous random variable $X$ describes how probability density is distributed over possible values of $X$.

- The joint PDF of two continuous random variables $X$ and $Y$ describes how density is distributed over pairs $(x,y)$

If $X$ and $Y$ are continuous random variables, the function
$$f_{XY}(x,y)$$ 
is called the joint probability density function of $X$ and $Y$, where probabilities are obtained by integration:
$$
P(a \leq X \leq b, c \leq Y \leq d) = \int_a^b \int_c^d f_{XY}(x,y) \: dy \ dx
$$

It has the following properties:
- $f_{XY}(x,y) \geq 0$
- $\int_{-\infty}^{\infty} \int_{-\infty}^{\infty} f_{XY}(x,y) \: dx \ dy = 1$
- $P((X,Y) \in R) = \iint_R f_{XY}(x,y) \: dy \ dx$

### Continuous multivariate joint distributions

- If $X_1, X_2,\ldots, X_n$ are all continuous random variables, the joint PDF is the function $f(x_1, x_2, \ldots, x_n)$ such that for any $n$ intervals $[a_1, b_1], \ \ldots, \ [a_n, b_n]$:

$$
P(a_1 \leq X_1 \leq b_1, \ \ldots, \ a_n \leq X_n \leq b_n) = \int_{a_1}^{b_1} \ldots \int_{a_n}^{b_n} f(x_1, x_2, \ldots, x_n) \: dx_n \ldots dx_1
$$

### Continuous marginal probability distributions

For a continuous joint distribution, a distribution of a single random variable can by integrating over the other variable(s):

- Let $X$ and $Y$ be two continuous random variables with joint probability density function $f_{XY}(x, y)$. Then, the corresponding marginal PDFs are given by:
$$
f_X(x) = \int f_{XY}(x,y) \: dy
$$
$$
f_Y(y) = \int f_{XY}(x,y) \: dx
$$ 

### Continuous conditional distributions

Let $X$ and $Y$ be two continuous random variables with joint probability density function $f_{XY}(x, y)$.

Assume $f_X(x) > 0$ (meaning $X = x$ is possible), then:

$$
f_{Y|X=x}(y) = \frac{f_{XY}(x, y)}{f_X(x)}
$$

Similarly, if $f_Y(y) > 0$:

$$
f_{X|Y=y}(x) = \frac{f_{XY}(x, y)}{f_Y(y)}
$$

The conditional distribution is a valid PDF, so:
1. $\int_{- \infty}^{\infty} f_{Y|X=x}(y) \: dy = 1$

2. For any set $B \subseteq \mathbb{R}$:
$$
P(Y \in B \mid X = x) = \int_{B} f_{Y|X=x}(y) \: dy
$$

### Independence and dependence

Let $X$ and $Y$ be two random variables (discrete or continuous) with possible values $\mathcal{X}$ and $\mathcal{Y}$ respectively.  
They are independent if the following equivalent statements hold:

- $f_{XY}(x,y) = f_X(x)\,f_Y(y)$ for all  $x \in \mathcal{X},\; y \in \mathcal{Y}$
- $f_{Y \mid X = x} = f_Y(y)$ for all $x,y$ with  $f_X(x) > 0$
- $f_{X \mid Y = y} = f_X(x)$ for all $x,y$ with  $f_Y(y) > 0$
- $P(X \in A,\; Y \in B) = P(X \in A)\,P(Y \in B)$ for all sets  $A, B$

### Mean and variance of joint random variables

To find the mean or variance of a single random variable in a joint distribution, first obtain its marginal distribution, then compute the expectation or variance as in the standard, single-variable case.

#### Properties of mean and variance

For any random variables $X$ and $Y$ and constants $a$, $b$, and $c$:
- $E[a] = a$

- $E[X + Y] = \mu_X + \mu_Y$

- $E[aX + bY + c] = a\mu_X + b\mu_Y + c$

- $E[h(X,Y)] = \sum_x \sum_y h(x,y)\, f_{XY}(x,y)$ (discrete)

- $E[h(X,Y)] = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} h(x,y)\, f_{XY}(x,y)\, dx\, dy$ (continuous)

- $V(X) = E[X^2] - \mu_X^2$

- $V(X + Y) = \sigma_X^2 + \sigma_Y^2 + 2\sigma_{XY}$

- $V(aX + bY + c) = a^2 \sigma_X^2 + b^2 \sigma_Y^2 + 2ab\,\sigma_{XY}$

### Covariance

Covariance measures the direction and strength of the linear relationship between two variables.

The covariance between two discrete or continuous random variables $X$ and $Y$ is denoted by $\text{Cov}(X, Y)$ and is given by:
$$
\text{Cov}(X, Y) = \sigma_{XY} = E[XY] - E[X]E[Y]
$$
Alternatively,
$$
\text{Cov}(X, Y) = E[(X - \mu_X)(Y - \mu_Y)]
$$

where $\mu_X = E[X]$ and $\mu_Y = E[Y]$. 

#### Properties of covariance

For any random variables $X$ and $Y$ and constants $a$, $b$, and $c$:

- $\text{Cov}(aX + b,\; cY + d) = ac\,\text{Cov}(X, Y)$

- $\text{Cov}(X_1 + X_2,\; Y) = \text{Cov}(X_1, Y) + \text{Cov}(X_2, Y)$

- $\text{Cov}(X, X) = \text{Var}(X)$

If $X$ and $Y$ are independent then $\text{Cov}(X,Y) = 0$

- The converse is not necessarily true - if covariance is $0$ the variables might not be independent

## Correlation coefficient

The correlation coefficient also measures the strength and direction of the linear relationship between two random variables, but in a standardised way. It is the normalised covariance.

The correlation coefficient between two random variables $X$ and $Y$ is:
$$
\text{Corr}(X, Y) = \rho_{XY} = \frac{\text{Cov}(X,Y)}{\sqrt{V(X)V(Y)}}
$$

It satisfies:
$$
-1 \leq \rho_{XY} \leq 1
$$

- $\rho = 1$ if and only if $Y = aX + b$ with $a > 0$
-  $\rho = -1$ if and only if $Y = aX + b$ with $a < 0$

If $X$ and $Y$ are independent random variables then they are uncorrelated, i.e., $\rho_{XY} = 0$

- the converse is not necessarily true - zero correlation need not imply independence
