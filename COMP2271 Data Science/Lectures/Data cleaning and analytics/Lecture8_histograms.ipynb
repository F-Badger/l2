```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

testdict={
                'key1': ['a','b','a','a','a','a','a','b'],
                'key2': ['c','c','c','c','d','d','d','d'],
                'num1':[1,1,1,4,4,6,7,8],
                'num2':[40,35,30,25,20,15,10,5]

}

pdtest=pd.DataFrame(testdict)
pdtest
```

```python
sns.displot(data=pdtest,x='key1')
```

```python
sns.displot(data=pdtest,x='key1',stat='probability')
```

```python
sns.displot(data=pdtest,x='num1',bins=[1,4,6,7,8])
```

```python
sns.displot(data=pdtest,x='num1',bins=[1,4,6,7,8],hue='key1',multiple='stack')
```
