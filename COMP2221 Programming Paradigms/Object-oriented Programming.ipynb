# Java

Java is a high-level, object-oriented programming language. 

It follows the principle of Write Once, Run Anywhere (WORA), meaning that Java programs are compiled into bytecode that can run on any system with a compatible Java virtual machine (JVM). It is commonly used in web applications, mobile development, and enterprise software.

## Java as an object-oriented language

Java is based on four key object-oriented programming (OOP) principles:

- Encapsulation
- Inheritance
- Polymorphism
- Abstraction

Everything in Java is an **object**:

- Objects contain attributes (data/state) and methods (behaviour)

A **class** is a blueprint or template that defines the fields and methods that its objects will have.

- This promotes modular, reusable, and maintainable code

## Java virtual machine

The Java virtual machine (JVM) is a virtual machine that executes Java bytecode. It:

- provides a platform-independent runtime environment
- translates bytecode into machine code for the host system
- enables Java applications to run on different operating systems

## Classes overview

A class is a template for creating objects: 

- it defines state (variables) and behavior (methods)
- a class does not occupy memory until an object is instantiated 

An example of a class is given below:

```java
class Car {
    // variables
    String brand;
    int speed;
    
    // methods
    void accelerate() {
    speed += 10;
    }
}
```

## Objects overview

An object is an instance of a class:

- it represents a specific realisation of the class
- it has state (attributes) and behaviour (methods)
- objects are stored in heap memory at runtime

An example of creating an object from the `Car` class is given below:

```java
public class TestCar {
    public static void main(String[] args) {
        Car myCar = new Car();
        myCar.brand = "Toyota";
        myCar.accelerate();
        System.out.println("Brand: " + myCar.brand + " Speed: " + myCar.speed);
    }
}
```

## Overview of OOP principles

### Encapsulation overview

Encapsulation means hiding internal details and exposing only what is necessary:

- private variables cannot be accessed directly from outside the class and are accessed via public methods (getters and setters)

```java
class Person {
    private String name;
    
    public String getName() {
        return name;
    }
    
    public void setName(String newName) {
        name = newName;
    }
}
```

### Inheritence overview

Inheritance allows a new class to inherit from an existing class: 

- this promotes code reusability and hierarchy

```java
class Animal {
    void makeSound() {
        System.out.println("Animal makes a sound");
    }
}

class Dog extends Animal { 
    // the Dog class inherits any attributes
    // and methods defined in the Animal class
    void makeSound() {
        System.out.println("Dog barks");
    }
}
```

### Polymorphism overview

Polymorphism allows a method to take different forms:

- method overriding is when a subclass provides a different implementation of a method from the parent class
- method overloading is when multiple methods have the same method name but different parameter lists (signature), possibly with different return behaviour

```java
class Animal {
    void makeSound() {
        System.out.println("Animal makes a sound");
    }
}

class Dog extends Animal {
    void makeSound() {
        // the Dog class overrides the makeSound()
        // function that it inherited
        System.out.println("Dog barks");
    }
}
```

### Abstraction

Abstraction hides implementation details and only shows relevant functionality:

- defines a template that enforces certain methods while hiding implementation details
- achieved using abstract classes or interfaces

```java
abstract class Animal {
    // No implementation
    abstract void makeSound();
}

class Cat extends Animal {
    void makeSound() {
        System.out.println("Meow");
    }
}
```

## Java basics

### Hello world

To create a basic Java program:

- save the code in a `.java` file
- the file may contain a class (only one public class allowed per file)
- if a class is declared `public`, the file name must match the class name
- the program must contain a `main` method
  - defined as `public static void main(String args[])`
- the JVM starts program execution from the `main` method
 
The implementation of 'Hello World' is given below:

````java
// MyMain.java
class MyMain {
    public static void main(String args[]){
        System.out.println("Hello World!");
    }
}

// Normally, Java programs run automatically 
// by the JVM starting from main.
// In Jupyter notebooks, 
// the main method must be called manually:
MyMain.main(new String[]{});
````

### Data types

Data types are split into primitive and non-primitive types.

Java has 8 primitive data types:

- `byte`
- `short`
- `int`
- `long`
- `float`
- `double`
- `char`
- `boolean`

Non-primitive types include:

- `String`
- Arrays
- Classes

### Declaration and assignment

Every variable in Java must have a declared data type before it is used.

The syntax for declaring a variable is:
```java
type variableName;
```

e.g.
```java
int x;
String y;
```

Values can be assigned using `=`:
```java
x = 5;
y = "ABC";
```

Declaration and assignment can be done in a single statement:
```java
int x = 5;
```

### Semi-colons

Most executable statements in Java must end in a semicolon (`;`).

e.g.,
```java
int x = 5;
System.out.println(x);
```

However, semicolons are not used after:

- class declarations
- method declarations
- control structures like `if`, `for`, and `while`

### Branching

Java supports conditional branching using `if`, `else if`, and `else` statements.

Branching allows a program to make decisions and execute different code depending on conditions.

The syntax is:

```java
if (condition) {
    // code
}
else if (condition) {
    // code
}
else {
    // code
}
```

The `else if` and `else` clauses are optional. 

An example use is:

````java
int a = 10;

if (a < 0) {
    System.out.println("Negative");
}
else {
    System.out.println("Positive");
}
````

### Loops 

Loops allow repeated execution of a block of code.

#### `for` loops

`for` loops are used when the number of iterations is known:

```java
for (int i = 0; i < 5; i++) {
    System.out.println("i=" + i);
}
```

- `continue` skips the rest of the current iteration and moves to the next iteration.
- `break` can be used to exit the loop.

#### `while` loops

`while` loops repeat execution while a condition is true:

```java
int j = 0;
while (j < 10) {
    System.out.println("j=" + j);
    j += 1;
}
```

- `break` can be used to exit the loop.

## Classes

A class is typically made up of three main components:

- Fields (instance variables) - store the object's state (data)
- Methods - define the object's behaviour
- Constructors - initialise new objects

An example is given below:
```java
class Person {
    // fields
    private String name;
    private int age;

    // constructor
    public Person(String name, int age) {
        this.name = name;
        this.age = age;
    }

    // class methods
    // getter for name
    public String getName() { return name; }

    // setter for name
    public void setName(String name) {
         this.name = name;
    }

    // getter for age
    public int getAge() { return age; }

    // setter for age
        public void setAge(Int age) {
         this.age = age;
    }

    // method to display stored information
    public void displayInfo() {
        System.out.println("Name: " + name +
                           ", Age: " + age);
    }
}
```

In this example:

- the fields are `private`
  - they cannot be accessed directly from outside the class
- the constructor and methods are public
   - they can be called from other classes

### Constructors
A constructor:

- must have the same name as the class
- must be `public`
- has no return type (not even `void`)
- is called automatically when an object is created with `new`
- initialises the object's fields

#### Multiple constructors

Constructors can be overloaded, meaning a class can define multiple constructors with different parameter lists.

This is an example of method overloading, a form of polymorphism.

The constructor that is called depends on the arguments provided when the object is created.

For example, in addition to 
```java
public Person(String name, int age) {
    this.name = name;
    this.age = age;
}
```

another constructor can be defined:

```java
public Person() {
    this.name = "No Name";
    this.age = 0;
}
```

This allows a `Person` object to be created without providing any arguments. In this case, the fields are initialised with default values chosen by the constructor. 

The same applies to other (non-constructor) methods:

- multiple implementations of the same method can be defined with different parameter lists, and the appropriate one is called depending on the arguments provided

## Creating an object

An object is created using the `new` keyword, which calls a constructor.
```java
ClassName objectName = new ClassName(arguments);
```

For example, for the `Person` class, objects can be created and initialised with:
```java
Person myPerson = new Person("Greg", 42);
// calls Person(String name, int age)

Person myUnnamedPerson = new Person();
// calls Person()
```

## Object and reference

When the following statement is executed:
```java
ClassName objectName = new ClassName(arguments);
```
the following happens:

- `ClassName objectName` declares a reference variable named `objectName` that references a `ClassName` object
- `new ClassName(arguments)` creates a new object in memory and calls the constructor
- the reference to the newly created object is assigned to `objectName`

Together, this statement creates an object and stores a reference to it in the variable.

### Reference copying

A reference variable can be copied so that two variables point to the same object.

```java
Person myPerson1 = new Person("Greg", 42);
Person myPerson2 = myPerson1;
```

Now both variables refer to the same object in memory. Therefore:

- if the object is modified through one reference, the change is visible through the other
