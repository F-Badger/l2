# Java

Java is a high-level, object-oriented programming language. 

It follows the principle of Write Once, Run Anywhere (WORA), meaning that Java programs are compiled into bytecode that can run on any system with a compatible Java virtual machine (JVM). It is commonly used in web applications, mobile development, and enterprise software.

## Java as an object-oriented language

Java is based on four key object-oriented programming (OOP) principles:

- Encapsulation
- Inheritance
- Polymorphism
- Abstraction

Everything in Java is an **object**:

- Objects contain attributes (data/state) and methods (behaviour)

A **class** is a blueprint or template that defines the fields and methods that its objects will have.

- This promotes modular, reusable, and maintainable code

## Java virtual machine

The Java virtual machine (JVM) is a virtual machine that executes Java bytecode. It:

- provides a platform-independent runtime environment
- translates bytecode into machine code for the host system
- enables Java applications to run on different operating systems

## Classes overview

A class is a template for creating objects: 

- it defines state (variables) and behavior (methods)
- a class does not occupy memory until an object is instantiated 

An example of a class is given below:

```java
class Car {
    // variables
    String brand;
    int speed;
    
    // methods
    void accelerate() {
    speed += 10;
    }
}
```

## Objects overview

An object is an instance of a class:

- it represents a specific realisation of the class
- it has state (attributes) and behaviour (methods)
- objects are stored in heap memory at runtime

An example of creating an object from the `Car` class is given below:

```java
public class TestCar {
    public static void main(String[] args) {
        Car myCar = new Car();
        myCar.brand = "Toyota";
        myCar.accelerate();
        System.out.println("Brand: " + myCar.brand + " Speed: " + myCar.speed);
    }
}
```

## Overview of OOP principles

### Encapsulation overview

Encapsulation means hiding internal details and exposing only what is necessary:

- private variables cannot be accessed directly from outside the class and are accessed via public methods (getters and setters)

```java
class Person {
    private String name;
    
    public String getName() {
        return name;
    }
    
    public void setName(String newName) {
        name = newName;
    }
}
```

### Inheritence overview

Inheritance allows a new class to inherit from an existing class: 

- this promotes code reusability and hierarchy

```java
class Animal {
    void makeSound() {
        System.out.println("Animal makes a sound");
    }
}

class Dog extends Animal { 
    // the Dog class inherits any attributes
    // and methods defined in the Animal class
    void makeSound() {
        System.out.println("Dog barks");
    }
}
```

### Polymorphism overview

Polymorphism allows a method to take different forms:

- method overriding is when a subclass provides a different implementation of a method from the parent class
- method overloading is when multiple methods have the same method name but different parameter lists (signature), possibly with different return behaviour

```java
class Animal {
    void makeSound() {
        System.out.println("Animal makes a sound");
    }
}

class Dog extends Animal {
    void makeSound() {
        // the Dog class overrides the makeSound()
        // function that it inherited
        System.out.println("Dog barks");
    }
}
```

### Abstraction overview

Abstraction hides implementation details and only shows relevant functionality:

- defines a template that enforces certain methods while hiding implementation details
- achieved using abstract classes or interfaces

```java
abstract class Animal {
    // No implementation
    abstract void makeSound();
}

class Cat extends Animal {
    void makeSound() {
        System.out.println("Meow");
    }
}
```

- if a class is declared `public`, the file name must match the class name
- the program must contain a `main` method

The implementation of 'Hello World' is given below:

````java
// MyMain.java
class MyMain {
    public static void main(String args[]){
        System.out.println("Hello World!");
    }
}

// Normally, Java programs run automatically 
// by the JVM starting from main.
// In Jupyter notebooks, 
// the main method must be called manually:
MyMain.main(new String[]{});
````

    // code

    // code

An example use is:

````java
int a = 10;

if (a < 0) {
    System.out.println("Negative");
}
else {
    System.out.println("Positive");
}
````

## Classes

A class is typically made up of three main components:

- Fields (instance variables) - store the object's state (data)
- Methods - define the object's behaviour
- Constructors - initialise new objects

An example is given below:
```java
class Person {
    // fields
    private String name;
    private int age;

    // constructor
    public Person(String name, int age) {
        this.name = name;
        this.age = age;
    }

    // class methods
    // getter for name
    public String getName() { return name; }

    // setter for name
    public void setName(String name) {
         this.name = name;
    }

    // getter for age
    public int getAge() { return age; }

    // setter for age
        public void setAge(Int age) {
         this.age = age;
    }

    // method to display stored information
    public void displayInfo() {
        System.out.println("Name: " + name +
                           ", Age: " + age);
    }
}
```

In this example:

- the fields are `private`
  - they cannot be accessed directly from outside the class
- the constructor and methods are public
   - they can be called from other classes

### `this`

The `this` keyword is used to refer to the current object in a method or constructor.

It is commonly used to eliminate confusion between class attributes and parameters with the same name.

### Constructors
A constructor:

- must have the same name as the class
- must be `public`
- has no return type (not even `void`)
- is called automatically when an object is created with `new`
- initialises the object's fields

Note that a constructor does not have to be explicitly provided. If one is not provided, one will be created automatically by the compiler. 

#### Multiple constructors

Constructors can be overloaded, meaning a class can define multiple constructors with different parameter lists.

This is an example of method overloading, a form of polymorphism.

The constructor that is called depends on the arguments provided when the object is created.

For example, in addition to 
```java
public Person(String name, int age) {
    this.name = name;
    this.age = age;
}
```

another constructor can be defined:

```java
public Person() {
    this.name = "No Name";
    this.age = 0;
}
```

This allows a `Person` object to be created without providing any arguments. In this case, the fields are initialised with default values chosen by the constructor. 

The same applies to other (non-constructor) methods:

- multiple implementations of the same method can be defined with different parameter lists, and the appropriate one is called depending on the arguments provided

## Creating an object

An object is created using the `new` keyword, which calls a constructor.
```java
ClassName objectName = new ClassName(arguments);
```

For example, for the `Person` class, objects can be created and initialised with:
```java
Person myPerson = new Person("Greg", 42);
// calls Person(String name, int age)

Person myUnnamedPerson = new Person();
// calls Person()
```

## Object and reference

When the following statement is executed:
```java
ClassName objectName = new ClassName(arguments);
```
the following happens:

- `ClassName objectName` declares a reference variable named `objectName` that references a `ClassName` object
- `new ClassName(arguments)` creates a new object in memory and calls the constructor
- the reference to the newly created object is assigned to `objectName`

Together, this statement creates an object and stores a reference to it in the variable.

### Reference copying

A reference variable can be copied so that two variables point to the same object.

```java
Person myPerson1 = new Person("Greg", 42);
Person myPerson2 = myPerson1;
```

Now both variables refer to the same object in memory. Therefore:

- if the object is modified through one reference, the change is visible through the other

## Encapsulation

**Encapsulation** in OOP is the bundling of data (fields) and methods that operate on that data into a single unit (a class), while restricting direct access to some of the object's internal state. 

The benefits of encapsulation are that it:

- protects data from accidental/erroneous modification
- enhances data hiding and security
- promotes modular and maintainable code

Java achieves this using access modifiers, mainly:

- `private`, and
- `public`

### Access modifiers

`private` and `public` are access modifiers that control how much of a class is visible to other parts of the program, allowing only necessary access.

They are used to ensure other classes access only the minimum information needed to perform their tasks.

`private` makes a variable/method accessible only within the same class. 

`public` makes a variable/method accessible from anywhere in the program. 

For a `private` field, `public` methods (called getters and setters) can be used to access and update the field, and can include validation to ensure that updates are valid.

```java
class Person {
    // both fields are private
    private String name;
    private int age;
    
    // setter for name
    public void setName(String newName) {
        this.name = newName;
    }
    
    // getter for name
    public String getName() {
        return name;
    }
    
    // setter for age (with validation)
    public void setAge(int newAge) {
        if (newAge > 0) {
            this.age = newAge;
        } else {
        System.out.println("Invalid age!");
        }
    }
    
    // getter for age
    public int getAge() {
        return age;
    }
}
```

## `static`

The `static` keyword in Java is used to create class-level members that belong to the class rather than an instance of the class:

- instance members belong to objects (each object gets its own copy)
- static members belong to the class itself (shared across all instances of the class)

```java
class ExampleClass {
    static int staticVar = 10;
    // shared across all objects
    
    int instanceVar = 20;
    // separate for each object
}
```

Only one copy of the `static` variable exists, regardless of how many objects are created. Therefore, updating the value for one object updates it for all objects of that class (since the variable is shared). 

Since `static` variables belong to the class, they can be accessed through an instance or directly through the class. However, the recommended way is using the class name. 

For example, `staticVar` can be accessed using
```java
ExampleClass.staticVar
```
without needing to create any instances of `ExampleClass`. 

````java
class Counter {
    static int staticCounter = 0;
    int instanceCounter = 0;
}

Counter counter1 = new Counter();
counter1.staticCounter ++;
counter1.instanceCounter ++;

Counter counter2 = new Counter();
counter2.staticCounter ++;
counter2.instanceCounter ++;

// staticCounter is shared
// Each object has their own instanceCounter
System.out.println(
    "counter1.staticCounter = " + counter1.staticCounter + 
    "\ncounter1.instanceCounter = " + counter1.instanceCounter + 
    "\ncounter2.staticCounter = " + counter2.staticCounter + 
    "\ncounter2.instanceCounter = " + counter2.instanceCounter
);
````

### `static final`

The `static final` combination is used for constants. These are variables that cannot be changed after initialisation.

```java
class Config {
    static final double PI = 3.14159;
}
```

By convention, constants are written in uppercase. 

### `static` methods

A `static` method belongs to the class, not to objects.

It cannot access instance (non-static) variables/methods directly:

- this is because static methods exist without any specific object instance
- a static method can therefore run without any object being guaranteed to exist

It can however use instance variables by passing an object (or values) as arguments:

```java
class Person {
    int age;

    static void printAgeFromObj(Person p) {
        System.out.println(p.age);
    }

    static void printAgeFromVal(int pAge) {
        System.out.println(pAge);
    }
}
```

## Inheritance

Inheritance is an OOP concept that allows a subclass (child class) to be derived from a superclass (parent class), inheriting its fields and methods and optionally adding new features or overriding existing behaviour.

Inheritance has many benefits:

- code reusability
  - shared behaviour is written once in the parent class
  - subclasses reuse it automatically
- easier maintenance
  - if the parent class is updated, all subclasses automatically get the update
- hierachy modelling
  - inheritance helps model real-world relationships
    - e.g., `Dog` is a subclass of `Animal`
  - this makes programs more natural and organised
- extensability
  - shared behaviour can remain in the parent class
  - specialised behaviour can be added in the subclass

In Java, the `extends` keyword is used to create a subclass:

```java
class SuperClass {
    // Parent properties and methods
}

class DerivedClass extends SuperClass {
    // Additional properties and methods
}
```

The subclass inherits all non-`private` members of the superclass. 

### Method overriding

Method overriding occurs when a subclass provides a new implementation of a method that is already defined in its superclass.

For a method to be overridden, the subclass method must have the same:

- method name
- parameter list
- return type

In Java, the `@Override` annotation is used to indicate that a method is intended to override a superclass method:

```java
class Animal {
    void makeSound() {
        System.out.println("Animal makes a sound");
    }
}
class Dog extends Animal {
    @Override void makeSound() {
        System.out.println("Dog barks");
    }
}
```

Note that using `@Override` is not strictly necessary, but is good practice. 

If `@Override` were removed from the example above, the program would still behave exactly the same.

`@Override` should be used because:

- it makes it clear to readers that the method is overriding a superclass method
- the compiler checks that the method actually overrides a valid method
- it helps catch mistakes such as:
  - typos in the method name
  - incorrect parameter lists
  - incorrect return types

Without `@Override`, if the method signature does not match the superclass method, Java treats it as a new overloaded method rather than an overridden one.

### `super`

The `super` keyword is used to refer to the immediate superclass (parent class) of the current object.

It is commonly used to:

- call a superclass constructor using `super(arguments)`
- call a superclass method using `super.methodName(arguments)`
- access a superclass field using `super.fieldName`

#### Calling a superclass constructor

A subclass constructor must call a superclass constructor.:

- if the constructor does not explicitly call `super(...)`, Java automatically inserts a call to `super()` as the first statement
- therefore, if the superclass does not have a no-argument constructor, the subclass must explicitly call one of the superclass constructors
- if used, `super(...)` must be the first statement in the subclass constructor.

```java
class Animal {
    String name;

    Animal(String name) {
        this.name = name;
    }
}

class Dog extends Animal {
    Dog(String name) {
        super(name);   // calls Animal(String name)
    }
}
```

#### Calling a superclass method

`super.methodName(arguments)` calls the superclass version of a method, even if it has been overridden in the subclass.

````java
class Animal {
    void makeSound() {
        System.out.println("Animal sound");
    }
}

class Dog extends Animal {
    @Override
    void makeSound() {
        super.makeSound(); // calls Animal.makeSound()
        System.out.println("Bark");
    }
}

Dog myDog = new Dog();
myDog.makeSound();
````

### `protected`

The `protected` access modifier allows a variable or method to be accessed in a more flexible way than private, but still restricts full public access.

A protected member is accessible:

- within the same class
- within the same package
- in subclasses 

It is not accessible from unrelated classes outside the package.

## Abstract classes

Abstract classes are declared using the `abstract` keyword and are used as base classes that cannot be instantiated directly (they can only be used as superclasses). 

They may contain:

- abstract methods (methods without a body)
- concrete methods (methods with a body)
- fields (variables)

Abstract methods:

- are also declared using the `abstract` keyword
- do not have an implementation 
- must be implemented by concrete (non-abstract) subclasses

```java
abstract class Animal {
    abstract void makeSound();
}
class Dog extends Animal {
    void makeSound() {
        System.out.println("Dog barks");
    }
}
```

## Interfaces

An interface in Java defines a contract that classes must follow. It specifies what a class must do, but not how it does it.

- Interfaces are declared using the `interface` keyword.
- Before Java 8, interfaces contained only abstract methods. After Java 8, non-abstract methods are also allowed.
- They define behaviours that a class must implement, and support multiple inheritance (meaning a class can inherit multiple interfaces)
  - this is a key difference between interfaces and abstract classes; a class cannot inherit multiple superclasses

A class uses the `implements` keyword to follow an interface.

- it must provide implementations for all abstract methods

```java
interface ElectricVehicle {
    void chargeBattery();
}

interface Vehicle {
    void start();
    void stop();
}

class ElectricCar implements Vehicle, ElectricVehicle {
    public void start() {
        System.out.println("Electric car starts silently.");
    }
    public void stop() {
        System.out.println("Electric car is stopping.");
    }
    public void chargeBattery() {
        System.out.println("Charging battery...");
    }
}
```

![UML dependency](uml_dependency.png)

## Abstraction

Abstraction in Java allows a superclass or interface to define the structure (method signatures) while leaving the actual implementation to subclasses:

- the superclass implements only the high-level design (e.g. method signatures for abstract methods) without knowing the specific logic, without implementing the specific logic, which becomes the responsibility of the subclass
- abstract classes and interfaces specify what operations are available, without defining how they are carried out

This ensures that different objects can share a common interface (behaviour, standard) without revealing the implementation details.

### Abstraction vs encapsulation

Abstraction:

- Defines what a class should do but not how
- Hides implementation details by providing method signatures without actual implementations in abstract classes or interfaces.
  - the child class must implement the details
- Focuses on reducing complexity by showing only essential behaviour

Encapsulation:
- Hides internal data and restricts access to it using access modifiers (`private`, `protected`, `public`).
- Ensures that only necessary parts of an object are accessible from outside.
- Focuses on protecting and controlling access to an object’s state

### Benefits of abstraction

Using abstraction:

- reduces complexity by hiding implementation details and exposing only essential behaviour
- enhances code reusability by allowing different classes to share the same interface or abstract structure
- improves maintainability because changes to implementation do not affect code that depends on the abstraction (as long as the interface remains the same)

### Implementing abstraction

In Java, abstraction is implemented using:

- [abstract classes](#Abstract-classes)
- [interfaces](#Interfaces)

Before Java 8, abstract classes and interfaces were quite different, as interfaces could only contain method declarations and constants. Since Java 8, interfaces can also include concrete implementations through `default` and `static` methods, making them more similar to abstract classes.

However, important differences remain:

- abstract classes can contain instance variables, constructors, and methods with any access modifier
- interfaces cannot have instance variables or constructors, and their methods are public by default
- a class can extend only one abstract class, but it can implement multiple interfaces

### `static` interface methods

Interfaces can contain `static` methods:

- these are called using the interface name, not through an object of an implementing class
- they are not inherited by `implenting` classes
- they are typically used for utility methods related to the interface

### `default` interface methods

Interfaces may also contain `default` methods:

- these are declared using the `default` keyword and include a method body
- if no keyword is specified, the method is implicitly `public abstract`

`default` methods:

- are inherited by implementing classes
- can be overridden by implementing classes
- if two or more interfaces provide a default implementation for the same method, the implementing class must provide its own implementation to avoid
ambiguity

### Interfaces as a type

Interfaces can be used as a type for variables, method parameters, and return types.

This allows code to work with any object that implements the interface, rather than depending on a specific class.

For example, if a method parameter has an interface type, the method can accept any object whose class implements that interface.

````java
interface Animal {
    void sound();
}

class Dog implements Animal {
    public void sound() {
        System.out.println("Woof");
    }
}

static void makeSound(Animal a) {
    a.sound();
    // any variable that implements the
    // Animal interface must implement
    // the sound() function
}

Dog myDog = new Dog();
myDog.sound()
````

### Multi-inheritance

Multiple inheritance means that a class inherits from two or more parent classes.

Java does not allow multiple inheritance of classes:

- Java supports only single inheritance for classes
- a class can extend only one superclass
- however, a class can implement multiple interfaces

This restriction avoids the diamond problem:

- if a class inherits from two superclasses that both provide different implementations of the same method, it becomes ambiguous which implementation should be inherited

Interfaces help avoid this issue:

- traditionally, interfaces contained only method declarations, so there was no ambiguity because the implementing class provided the implementation
  - since Java 8, interfaces can include `default` methods
- if two interfaces provide the same default method, the implementing class must override the method to resolve the conflict explicitly

### When to use abstract classes or interfaces

Abstract classes should be used when:

- code (variables and methods) needs to be shared among closely related classes
  - especially when sharing state (instance variables), since interfaces cannot have instance fields
- there is partially implemented behaviour that subclasses will extend or customise
- classes share a strong 'is-a' relationship and a common base implementation is useful

Interfaces should be used when:

- a contract needs to be enforced between potentially unrelated classes
- multiple inheritance of type is required
- you want to define behaviour without relying on shared state
- default behaviour is needed via `default` methods, but without shared instance variables

## Polymorphism

Polymorphism means 'many forms'.

In OOP, it allows a single “thing” (object or method) to behave in different ways depending on context (different classes or implementations).

It is achieved through:

- Method overloading (compile-time polymorphism)
- Method overriding (run-time polymorphism)
- Upcasting and downcasting (type conversion in inheritance)

Polmorphism improves reusability, flexibility and maintainability.

### Method overloading (compile-time polymorphism)

Method overloading is when multiple methods have the same name but different parameter lists. It happens within the same class. 

The compiler determines which method is called at compile-time based on the method signature.

Method overloading was covered earlier, with an example, in the [Classes](#Multiple-constructors) section. 

### Method overriding (run-time polymorphism)

Method overriding allows a subclass to provide a specific implementation of a method already defined in its superclass. 

The method that is executed is determined at run-time based on the actual object type (not the reference type).

Method overriding was covered earlier, with an example, in the [Inheritance](#Method-overriding) section. 

### Upcasting

Upcasting is when a subclass object is assigned to a superclass reference.

Only methods defined in the superclass are accessible. However any overridden methods will still use the subclass implementation (run-time polymorphism).

````java
class Animal {
    void makeSound() { System.out.println("Animal sound."); }
}

class Dog extends Animal {
    void makeSound() { System.out.println("Bark!"); }
    void walk() { System.out.println("Walking..."); }
}

Animal myPet = new Dog(); // upcasting
// a Dog object is referenced as an Animal

myPet.makeSound();
// calls Dog's overridden method at runtime

// myPet.walk(); 
// not accessible (not defined in Animal)
````

#### Upcasting in method parameters

Upcasting allows a method that takes a superclass parameter to accept objects of any subclass.

The subclass object is implicitly upcast to the superclass type when passed as an argument.

````java
Dog myDog = new Dog(); 

void makeSound(Animal a) {
    // expects object with class Animal
    a.makeSound();
}

makeSound(myDog);
````

#### Upcasting in arrays and collections

Upcasting also allows arrays and collections to store objects of different subclasses, as long as they share a common superclass.

````java
class Cat extends Animal  {
    void makeSound() { System.out.println("Meow!"); }
}

class Cow extends Animal  {
    void makeSound() { System.out.println("Moo!"); }
}

Cat myCat = new Cat();
Cow myCow = new Cow();

Animal[] zoo = {myDog, myCat, myCow};
// array of Animal objects
// each object is implicitly upcast to Animal

for (int i = 0; i < zoo.length; i++){
    zoo[i].makeSound();
}
````

### Downcasting

Downcasting is when a superclass reference is converted (typecasted) back to a subclass reference:

- it must be performed explicitly using a cast
- it allows subclass-specific methods and fields to be accessed
- it is only safe if the object is actually an instance of the subclass

````java
Animal myPet = new Dog(); // upcasting
Dog myDog = (Dog) myPet; // downcasting

myDog.walk();
// can use methods defined in Dog
// but not in Animal
````

Downcasting only works when for reverting an object back to a class it was originally upcast from. 

### `instanceof`

The `instanceof` is a binary operator thattests whether an object is an instance of a given class or interface.

It returns `true` if the object is of the specified type (or a subtype), and `false` otherwise.

````java
Animal myPet = new Dog(); 

System.out.println(myPet instanceof Animal);
System.out.println(myPet instanceof Dog);
System.out.println(myPet instanceof Cat);
System.out.println(myPet instanceof Cow);
````

`instanceof` can be used to check whether a downcast is safe before performing it.

````java
void takeForWalk(Animal a) {
    if (a instanceof Dog) {
        Dog myDog = (Dog) a;
        myDog.walk();
    }
    else {
        System.out.println("Cannot walk");
    }
}

Animal myPet = new Dog();
System.out.println("myPet:");
takeForWalk(myPet);

Animal myPet2 = new Cow();
System.out.println("myPet2:");
takeForWalk(myPet2);
````

If the compiler can determine that an `instanceof` test can never be `true`, it reports a compile-time error.

This happens when the two types are unrelated, so no object could possibly be an instance of both.

For example,
```java
Cat myCat = new Cat();
System.out.println(myCat instanceof Dog);
// produces compile time error
// Cat and Dog classes are unrelated
// an object that can be a Cat can never be a Dog
```

## Generics

Generics enable types to be used as parameters. This makes it possible to write code for different data/object types using a single class or method while maintaining type safety.

The angle brackets `<T>` is used to denote a generic type parameters. Multiple generic types can be used, such as `<T, U, V, ...>`.

### Generic class

The example below is for a class that works with different data types using generics.

````java
class Box<T> {
    T value; // T is a placeholder for any data type
    
    void setValue(T value) {
        this.value = value;
    }

    T getValue() {
        return this.value;
    }
}

// Create a Box to hold a String
Box<String> stringBox = new Box<>();
stringBox.setValue("Hello");
System.out.println("Value: " + stringBox.getValue());

// Create a Box to hold an Integer
Box<Integer> intBox = new Box<>();
intBox.setValue(50);
System.out.println("Value: " + intBox.getValue());
````

### Generic method

It is also possible to create methods that work with any data type using generics:

````java
// Generic method: works with any type T
public static <T> void printArray(T[] arr) {
    for (int i = 0; i < arr.length; i++) {
      System.out.println(arr[i]);
    }
}

// Array of Strings
String[] names = {"Eric", "Robert"};

// Array of Integers
Integer[] numbers = {1, 2, 3};

// Call the generic method with both arrays
printArray(names);
printArray(numbers);
````

### Primitive types in generics

Primitive types (`int`, `float` etc.) cannot be type parameters.

- instead, wrapper classes (`Integer`, `Float` etc.) must be used.

### Type erasure

During compilation, type parameters are replaced with the `Object` type (for unbounded generics).

- `Object` is the root superclass of all classes
- it contains contains basic methods that all classes (e.g. `toString()`)

```java
class Box<T> {
    T value;
}
```
becomes
```java
class Box {
    Object value;
}
```

Therefore, in a method or class that uses unbounded generics, only operations available to `Object` are allowed.

### Bounded generics

Type parameters can be restricted to only allow types that extend a particular class or implement a partiular interface. 

```java
<T extends Number>
```
This means `T` can only be `Number` or any subclass of `Number` (e.g. `Integer`, `Double`, etc.).

Now, the type parameter is not replaced with `Object`.

Instead, it is replaced with the upper bound (`Number` in this case).

```java
class Box<T extends Number> {
    T value;
}
```
becomes
```java
class Box {
    Number value;
}
```

Therefore, methods defined in `Number` are allowed.

## Collections

The Java Collections Framework (JCF) is a unified architecture for handling and manipulating groups of objects efficiently.

It provides a set of interfaces and classes for storing, retrieving, and processing data dynamically.

It offers many benefits:

- reusability - provides standardised, well-tested data structures and algorithms
- performance - includes optimised implementations for operations like sorting and searching
- flexibility - supports dynamic resizing (unlike arrays, which are fixed size)
- thread safety - some collections support concurrent access
- reduced boilerplate code - removes the need to implement common data structures manually

Items in the JCF need to be imported before they can be used. They are imported from `java.util`. 

```java
import java.util.*;
```
This imports all classes in the `java.util` package. 

Alternatively, specific classes can be imported. For example:
```java
import java.util.ArrayList;
```
This imports only the `ArrayList` class.

### Arrays vs collections

#### Arrays
In Java, an array can be declared using:

```java
typeName[] arrayName = new typeName[arrayLength];
```

For example,
```java
int[] nums = new int[5];
```
creates an array of `int`s called `nums`, with length 5. 

Arrays:

- have fixed size (cannot grow or shrink after creation)
- require manual handling for insertions and deletions (shifting elements)
- have no built-in methods for sorting, searching, or managing data

#### Collections

A collection can be declared using:
```java
List<typeParameter> collectionName = new ArrayList<>();
```

For example, 
```java
List<Integer> nums = new ArrayList<>();
``` 
creates a dynamically sized list of integers called `nums`.

An element can be accessed by index using
```java
collectionName.get(index)
```

Collections:

- support dynamic resizing (can grow and shrink at runtime)
- provide built-in methods such as `add()`, `remove()`, `contains()`, `sort()`
- use generics to enforce type safety (e.g. `List<Integer>`)
- can store objects (primitive types are auto-boxed, e.g. `int` → `Integer`)

### Java collections categories

The Java Collections Framework is divided into four main categories:
1. `List`: an ordered collection that allows duplicates
2. `Set`: a collection that does not allow duplicate elements
3. `Map`: stores key-value pairs with unique keys
4. `Queue`: first-in-first-out FIFO structure

### `LinkedList`

In addition to `ArrayList`, the JCF provides a `LinkedList`:

- elements are stored in nodes (containers)
- a `LinkedList` is a doubly-linked list:
  - each node contains a reference to the next node and the previous node
  - the list maintains references to the first (head) and last (tail) nodes

### `ArrayList` vs `LinkedList`

Each has relative advantages:

- it is faster to access random elements in an `ArrayList`
  - elements can be directly accessed by index in constant time
  - in a `LinkedList`, nodes must be traversed sequentially from the head (or tail) until the target is reached
- it is quicker to add or remove elements from a `LinkedList`
  - only the references of the adjacent nodes need to be updated
  - in an `ArrayList`, elements may need to be shifted in memory to maintain contiguous storage

### Looping over lists

Arrays, `ArrayList`s and `LinkedList`s can be iterated using a `for` loop.

This can be done by iterating over each index until the size of the list is reached:

````java
List<String> names = new ArrayList<>();

names.add("Robert");
names.add("Allison");
names.add("Eric");

for (int i = 0; i < names.size(); i++) {
    System.out.println(names.get(i));
}
````

#### Enhanced for loop (for-each loop)
The enhanced for loop can also be used, which directly accesses each element in the list:
```java
for (typeName element : listName) {
    // code
}
```

For example:

````java
for (String name : names) {
    System.out.println(name);
}
````

### `Iterator`s

The `Iterator` class from the JCF can also be used to loop over a collection such as a list.

An iterator is created using:

```java
Iterator<typeName> iteratorName = listName.iterator();
```

Then, a while loop can be used with the `hasNext()` and `next()` methods to traverse the collection.

- `hasNext()` returns `true` if there is another element, otherwise false
- `next()` returns the next element in the iteration and advances the iterator

For example, 

````java
Iterator<String> iterNames = names.iterator();

while (iterNames.hasNext()) {
    System.out.println(iterNames.next());
}
````

`Iterator`s are useful because:

- they allow removal of elements while traversing the collection using `iterator.remove()`
  - a `for` loop does not
- `Iterator`s can be used for most collection types (e.g. `List`, `Set`, etc.)
- the specific class `ListIterator` can be used for bidirectional traversal of lists (forward and backward)

An example with element removal is given below:

````java
Iterator<String> iterNames = names.iterator();

while (iterNames.hasNext()) {
    String c = iterNames.next();
    if (c.equals("Eric")) {
        iterNames.remove();
    }
}

System.out.println(names);
````

### `HashMap`

`HashMap` is a JCF class that:

- stores data as key-value pairs
- maps keys to values
  - keys must be unique
- uses hashing for fast lookups
- has no guaranteed order

A `HashMap` is created using:

```java
HashMap<KeyType, ValueType> hashMapName = new HashMap<KeyType, ValueType>();
```

A key-value pair can be added using:
```java
hashMapName.put(key, value)
```

A value can be retrieved using:
```java
hashMapName.get(key)
```

For example:

````java
HashMap<String, String> specialties = 
    new HashMap<String, String>();

specialties.put("Chase", "Intensivist");
specialties.put("Cameron", "Immunologist");
specialties.put("Foreman", "Neurologist");

System.out.println(specialties);
System.out.println(specialties.get("Foreman"));
````

### `HashSet`

`HashSet` is a JCF class that:

- stores unique elements only
- is backed internally by a `HashMap`
- does not preserve insertion order (order is not guaranteed)
- uses hashing for fast lookup

A `HashSet` can be created using 
```java
HashSet<typeName> setName = new HashSet<typeName>();
```

Elements can be added using
```java
setName.add(element);
```

To check if an element exists in the set, use:
```java
setName.contains(element); // returns a Boolean
```
