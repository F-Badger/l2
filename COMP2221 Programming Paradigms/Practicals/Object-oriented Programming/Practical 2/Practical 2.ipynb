````java
abstract class Animal {
    String name;

    public Animal(String name) {
        this.name = name;
    }

    abstract void makeSound();

    public void displayInfo() {
        System.out.println(
            "Animal name: "+ this.name
        );
    }
}

class Dog extends Animal {
    public Dog(String name){
        super(name);
    }
    
    void makeSound() {
        System.out.println("Auww");
    }
}

class Cat extends Animal {
    public Cat(String name){
        super(name);
    }
    
    void makeSound() {
        System.out.println("Meow");
    }
}

public class Main {
    public static void main(String[] args) {
        Dog myDog = new Dog("Jojo");
        Cat myCat = new Cat("Garfield");
        myDog.makeSound();
        myDog.displayInfo();
        myCat.makeSound();
        myCat.displayInfo();
    }
}

Main.main(new String[]{});
````

````java
abstract class Appliance {
    String brand;

    public Appliance(String brand) {
        this.brand = brand;
    }

    abstract void showFunction();

    public void displayBrand() {
        System.out.println(this.brand);
    }
}

interface SmartDevice {
    void connectToWiFi();
}

class SmartFridge extends Appliance implements SmartDevice {
    public SmartFridge(String brand) {
        super(brand);
    }
    public void showFunction() {
        System.out.println("Keeping stuff cool");
    }
    public void connectToWiFi() {
        System.out.println("Connected to WiFi");
    }
}

public class Main {
    public static void main(String[] args) {
        SmartFridge mySmartFridge = new SmartFridge("Bosch");
        mySmartFridge.displayBrand();
        mySmartFridge.showFunction();
        mySmartFridge.connectToWiFi();
    }
}

Main.main(new String[]{});
````

````java
abstract class Transport {
    String model;

    public Transport(String model) {
        this.model = model;
    }

    abstract void move();

    public void showModel() {
        System.out.println("Transport model: "+ this.model);
    }
}

interface SelfDriving {
    void autoPilot();
}

class AutonomousCar extends Transport implements SelfDriving {
    public AutonomousCar(String model) {
        super(model);
    }
    public void move() {
        System.out.println("Car is moving");
    }
    public void autoPilot() {
        System.out.println("Car is in autopilot");
    }
}

class AutonomousDrone extends Transport implements SelfDriving {
    public AutonomousDrone(String model) {
        super(model);
    }
    public void move() {
        System.out.println("Drone is moving");
    }
    public void autoPilot() {
        System.out.println("Drone is in autopilot");
    }
}

public class Main{
    public static void main(String[] args) {
        AutonomousCar myCar = new AutonomousCar("Car McCarface");
        AutonomousDrone myDrone = new AutonomousDrone("Pigeon");
        myCar.showModel();
        myCar.autoPilot();
        myCar.move();
        myDrone.showModel();
        myDrone.autoPilot();
        myDrone.move();
    }
}

Main.main(new String[]{});
````
