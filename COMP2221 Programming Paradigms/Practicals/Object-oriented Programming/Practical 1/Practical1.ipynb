````java
class Book {
    private String title;
    private String author;
    private double price;

    public String getTitle(){
        return this.title;
    }

    public String getAuthor(){
        return this.author;
    }

    public double getPrice(){
        return this.price;
    }

    public Book(String pTitle, String pAuthor, double pPrice) {
        this.title = pTitle;
        this.author = pAuthor;
        this.price= pPrice;
    }
}

class Main {
    public static void main(String[] args) {
        Book book1 = new Book("Book1", "Bob", 5.5);
        Book book2 = new Book("Book2", "John", 15.80);

        System.out.println(
            "Book 1:" +
            "\nTitle: " + book1.getTitle() +
            "\nAuthor: " + book1.getAuthor() +
            "\nPrice: " + book1.getPrice() + 
            "\n\nBook 2:" +
            "\nTitle: " + book2.getTitle() +
            "\nAuthor: " + book2.getAuthor() +
            "\nPrice: " + book2.getPrice()
        );
    }
}

Main.main(new String[]{});
````

````java
class BankAccount {
    static int totalAccounts = 0;
    private int accountNumber;
    private double balance;

    public BankAccount() {
        totalAccounts += 1;
        this.accountNumber = totalAccounts;
        this.balance = 0.0;
    }

    public void deposit(double amount) {
        balance += amount;
    }

    public double withdraw(double amount) {
        if (amount > balance) {
            double withdrawAmount = balance;
            balance = 0;
            return withdrawAmount;
        }
        else {
            balance -= amount;
            return amount;
        }
    }

    public void displayInfo() {
        System.out.println(
            "Account Number: " + accountNumber +
            "\nBalance: " + balance
            );
    }
    

    static void displayTotalAccounts() {
        System.out.println(
            "Total accounts: " + totalAccounts
            );
    }
}

public class Main {
    public static void main(String[] args) {
        BankAccount.displayTotalAccounts();
        BankAccount account1 = new BankAccount();
        BankAccount account2 = new BankAccount();

        account1.deposit(100);
        account1.displayInfo();

        account1.withdraw(75);
        account1.displayInfo();

        account2.deposit(50);
        account2.displayInfo();

        account2.withdraw(100);
        account2.displayInfo();

        BankAccount.displayTotalAccounts();
    }
}

Main.main(new String[]{});
````

### Solution 2: `BankAccount` Class

```java
class BankAccount {
    private static int totalAccounts = 0; // Shared across all accounts
    private int accountNumber;
    private double balance;
    
    public BankAccount(double initialBalance) {
        this.accountNumber = ++totalAccounts; // Assign a unique account number
        this.balance = initialBalance;
    }
    
    public void deposit(double amount) {
        if (amount > 0) {
            balance += amount;
        }
    }
    
    public void withdraw(double amount) {
        if (amount > 0 && amount <= balance) {
            balance -= amount;
        } else {
            System.out.println("Insufficient funds.");
        }
    }
    
    public void displayInfo() {
        System.out.println("Account Number: " + accountNumber +
                           ", Balance: $" + balance);
        }
    
    public static void displayTotalAccounts() {
        System.out.println("Total Accounts Created: " + totalAccounts);
    }
}
    
public static void main(String[] args) {
    BankAccount acc1 = new BankAccount(1000);
    BankAccount acc2 = new BankAccount(2000);
    acc1.displayInfo();
    acc2.displayInfo();
    BankAccount.displayTotalAccounts();
}
```
