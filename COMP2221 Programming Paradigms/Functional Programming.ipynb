# Functional programming

Functional programming is a programming paradigm where computation is done by applying functions to arguments (inputs). Programs specify what computations should take place (not necessarily how, as in imperative programming), with a focus on combining and chaining functions. It is similar in style to algebra. 

## Pure functions

Look at the Python code below. 

```python
state = 0

def f(n):
    global state
    state += 1
    return n + state
    
print(f(1))  # 2
print(f(1))  # 3
```

Here, making the same function call `f(1)` twice produces two distinct outputs. This is because `f` has a **side effect**:

- A side effect is any change a function makes to state or the outside world beyond returning its result.

A **pure function** is a function that takes all its inputs as arguments and produces all its outputs as results. It does not have any side effects. 

## Functional languages

A **functional programming language** is a programming language that supports and encourages the functional programming paradigm, where function application and composition are the basic building blocks. 

A language is **purely functional** if it forbids variable assignment and side effects, ensuring all functions are pure functions. This makes reasoning about the code simpler for both humans and the compiler. 

**Haskell** is an example of a purely functional language. 

### Example: computing $n!$

Using an imperative style, $n!$ may be calulcated using a for loop:

```python
def calculate_factorial(n):
    factorial = 1
    for i in range(1, n+1):
        factorial = factorial * i
    return factorial
```

However, this uses variable assignment (`factorial = factorial * i`), which is forbidden in a purely functional language. Instead, $n!$ can be calculated recursively, noting the piecewise representation of the factorial function: 
$$ 
F_n = 
\begin{cases}
    1 & n = 1 \\
    n \times F_{n-1} & \text{otherwise}
\end{cases}
$$

Pseudocode for this is given below:
```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
```

This can be implemented in Haskell using the code given below.

Note that comments in Haskell use `--`.

<a id="factorial-function"></a>

````haskell
factorial :: Int -> Int  -- declare factorial as a function that takes an int as input, and outputs an int
factorial 1 = 1  -- declare that factorial of 1 is 1
factorial n = n * factorial(n-1) -- declare that factorial of n is n * factorial of n-1

factorial 5  -- call the function with input value 5. Note that arguements do not need to be placed in brackets
````

## Types

A **type** is a collection of related values.

A type can be described by specifying:

- the set of values it contains, and
- the operations that can be performed on those values

For example, the `Bool` type contains the values `True` and `False`, and supports logical operations such as `&&` (and), `||` (or) and `not`. 

In Haskell, every expression (including functions) has a type. For example, `factorial :: Int -> Int` declares that the `factorial` function takes an `Int` as input and returns an `Int` as output.

Types are needed to:

- tell us how to interpret a variable
- define which operations are valid on a value
- help the compiler or interpreter find the correct implementation of an operator
- check if an operation on data is valid and well-defined (ensure code fragments are type-safe)
- document code clearly

### Type enforcement

Programs can be classified according to how strongly they enforce types:

- Strongly typed languages enforce types strictly
  - incompatible types cannot be mixed without explicit conversion
  - Haskell is a strongly typed language
- Weakly typed languages do not enforce types strictly
  - the language may automatically convert between types (type coercion)
  - Python is a weakly typed language

### Type checking

Translators must check for type correctness (verifying that operations are used with the correct types).   

Programming languages can be classified according to when they check for type correctness:

- Statically typed language: type safety is checked at translation time
  - invalid types result in translation error
  - Haskell is a statically typed language
- Dynamically typed language: type safety is checked at run time
  -  invalid types are only detected when they are used
  -  Python is a dynamically typed language

There are three main ways a translator can determine the type of an expression:

- Explicit annotation: the programmer annotates all variables with type information (e.g. C/Java)
- Type inference: the translator deduces types automatically from how variables are used (e.g. Haskell)
- Duck typing: the translator/runtime tries the operation, and if it succeeds, the type is valid (e.g. Python)

An example of type inference in Haskell is shown below.

````haskell
x :: Int
x = 6
y :: Int
y = 7
z = x + y
:t z  -- since x and y both have type Int, and z is the addition of these, the translator deduces that z must also be an Int
````

### Type notation in Haskell

In Haskell, two colons `::` are used to declare type: 

- `e :: T` means that `e` is of type `T`

It is possible to check the type of an expression using `:t expr` or `:type expr`. 

````haskell
x :: Int
x = 1
:t x
:type not  -- the not logical operator takes a Bool, and returns a Bool
````

## Currying
Currying is a way of writing functions so that they take one argument at a time and return a new function for the next argument.

- this means that multi-argument functions are actually a sequence of single-argument functions

Almost all functions in Haskell are written in the curried form. 
```Haskell
-- "uncurried"
add :: (Int, Int) -> Int  -- input is a tuple containing two Ints
add (x, y) = x + y

-- "curried"
add_curried :: Int -> Int -> Int  -- input is an Int, returns a function that takes
                                  -- another Int, which then returns an Int
add_curried x y = x + y
```

The `->` symbol associates to the right, removing the need for brackets:

- `Int -> Int -> Int -> Int` is equivalent to `Int -> (Int -> (Int -> Int))`

Function application associates to the left:

- `mult x y z` is equivalent to `((mult x) y) z`

## $\lambda$-calculus

$\lambda$-calculus is a formal system for describing computation using only:

- function definition (abstraction)
- function application

In $\lambda$-calculus:

- everything is a function
- there are no variables with changing state
- computation consists of applying functions to arguments

It forms the theoretical foundation of functional programming, and is the simplest known Turing-complete system (i.e., it can compute anything that any computer can in theory). 

### $\lambda$-expressions

$\lambda$-expressions can be defined inductively:
- A variable $v$ is a $\lambda$-expression
  - variables are represented by lower-case letters
- If $M$ is a $\lambda$-expression, then $(\lambda v . M)$ is a $\lambda$-expression
  - this defines a function:
    - $v$ is the parameter
    - $M$ is the function body
    - $\lambda$ is the start of the function definition
  - e.g. $\lambda x. x + 1$
- If $M$ and $N$ are $\lambda$-expressions, then $(MN)$ is a $\lambda$-expression
  - this means that the function $M$ is applied to the argument $N$
 
Note that functions take exactly one argument; currying is used to model multi-argument functions. 

Example of valid $\lambda$-expressions include:

- $x$: a variable
  
- $(\lambda x . x)$: the identity function (returns its input)

- $((\lambda x . x)\ e)$: the identity function applied to an expression $e$
  
- $(\lambda x . (\lambda y. (xy)))$: a (nested/curried) function that takes $x$ and returns another function that takes $y$ and applies $x$ to $y$

- $(((\lambda x . (\lambda y. (xy)))\ a)\ b)$: a nested function applied to expressions $a$ and $b$. This reduces to $(ab)$, i.e., the function $a$ applied to the argument $b$

### Associativity

Application is left-associative:

- $MNP$ means apply $M$ to $N$, and then apply that result to $P$, i.e., $((MN)P)$

Abstraction (function definition) is right-associative:

- $\lambda x . \lambda y . M$ means $\lambda x . (\lambda y . M)$

This eliminates the need to use brackets. 

Application takes precedence over abstraction:

- $\lambda x . \lambda y . xy$ is equivalent to $\lambda x . ( \lambda y . (xy))$, rather than $\lambda x . ( \lambda y . x)y)$


### Free vs. Bound variables

A variable is **bound** if it is a parameter of an enclosing function ($\lambda$-abstraction). 
- for instance, $x$ in $\lambda x . x$ is bound
- a variable binds to the closest enclosing $\lambda$-abstraction that declares the same variable name

A variable is **free** if it is not bound
- for instance, $x$ in $\lambda y . x$ is free (and $y$ is bound)

### Transformation rules

#### $\beta$-reduction

$\beta$-reduction is used to evaluate $\lambda$-expressions by substituting the argument of a $\lambda$-abstraction into its body:
$$
((\lambda x . M[x]) \ N) \;\to\; M[x := N]
$$

When an argument is applied (substitued), the outer $\lambda$-abstraction is eliminated.

Here:

- $\lambda x . M [x]$ is a $\lambda$-abstraction (function) with parameter $x$ and body $M[x]$ (like a function $f(x) = M[x]$)
- $N$ is the arguement being applied to the function
- $((\lambda x . M[x]) \ N)$ is applying the function to argument $N$
- $M[x:=N]$ is obtained by replacing every occurrence of the parameter $x$ in $M[x]$ with the actual argument $N$

For example:

- $(\lambda x . x) \ a = a$
  - replacing $x$ with $a$ in the body gives $a$
- $(\lambda x . y) \ a = y$
  - there is no occurrence of $x$ in the body, so the result is $y$

A more complex, nested example is:
$$ 
((\lambda x . (\lambda y . (xy))) \ a) \ b =
$$
$$
((\lambda y . (ay)) \ b =
$$
$$
ab
$$

A $\lambda$-expression is in $\beta$-normal form if no $\beta$-reduction is possible (the expression cannot be reduced further). 

#### $\alpha$-conversion

$\alpha$-conversion is used to rename bound variables in a $\lambda$-abstraction.

$$
(\lambda x . M[x]) \;\to\; (\lambda y . M[x := y])
$$

It replaces a bound variable with a fresh variable name.
- The structure and meaning of the expression remain unchanged.

This helps to avoid name conflicts and prevent variable capture during $\beta$-reduction:

- a name conflict happens when two different variables use the same name, even though they refer to different things
  - e.g., in the expression $\lambda x. (\lambda x. x)$, the outer $x$ and inner $x$ are different variables, but share the same name
- variable capture is when a free variable becomes bound accidentally during substitution ($\beta$-reduction)
  - e.g., take the expression $(\lambda x. (\lambda y. xy)) \ y$
    - subsituting $x := y$ would produce $\lambda y . yy$
    - the free $y$ (the argument) that was substituted in is now bound by $\lambda y$
    - the meaning of the expression has changed

$\alpha$-conversion renames the bound variable to prevent this from happening:

$$
(\lambda x. (\lambda y. xy)) \ y \;\to\; (\lambda x. (\lambda z. xz)) \ y
$$

Now, $\beta$-reduction produces:

$$
\lambda z . y z
$$

$y$ now stays free; the semantics of the expression is preserved. 

### $\lambda$-expressions in Haskell

As well as creating named functions as was done [earlier](#factorial-function), nameless functions can be constructed using $\lambda$-expressions. 

A $\lambda$-expression has the form:
```haskell
\x -> expression
```

Below is code for a nameless function that takes a number $x$ and returns $x + x$
```haskell
\x -> x + x
```
#### Use example 1
Unnamed functions can be used to formalise the idea of functions defined using currying:
```haskell
add x y = x + y
-- is equivalent to
add = \x -> (\y ->`` x + y)
```

The second form emphasises the idea that `add` is a function of one variable that returns a function (which takes the second variable). 

#### Use example 2
Unnamed functions can also be useful when returning a function as a result:
```haskell
const :: a -> b -> a
const x _ = x
```

The `const` function takes two inputs, and always returns the first, ignoring the second. This can be made much simpler using $\lambda$-notation:

```haskell
const :: a -> b -> a
const x = \_ -> x
```

This makes it clear that `const` takes its first argument and returns a function `\_ -> x`. This function can then take a second argument, which it ignores, and always returns the first argument `x`. 

#### Use example 3
Another use is for when a function is used only once. 

```haskell
odds :: Int -> [Int]
odds n = map f [0..n-1]
    where
        f x = x*2 + 1
```
This function generate the first `n` positive odd numbers.  

Using a $\lambda$-function, can simplify this function, removing the entire `where` clause:
```haskell
odds :: Int -> [Int]
odds n = map (\x -> x*2 + 1) [0..n-1]
```

### Translating between named functions and $\lambda$-expressions

Any multi-argument function can be rewritten using $\lambda$-expressions by moving arguments one-by-one to the right-hand side and wrapping each in a $\lambda$ function.

```haskell
f a b c = ...
f = \a b c -> ...
f = \a -> (\b -> (\c -> ...))
```

Both forms are equivalent; the choice of which to use is stylistic. 

## Lists

In Haskell, every non-empty list is built using the cons (construct) operator `:`. This operator takes an element and a list, and prepends the element to the list (adds it to the front of the list). 

Although lists are written just using square brackets, this is just syntactic sugar for repeated use of the `:` operator.  

```haskell
[1,2,3,4] 
-- is syntactic sugar for
1 : (2 : (3 : (4 : [])))
```

Elements can be accessed using the `!!` operator to index the array:

````haskell
x = [1,2,3,4]
x !! 2  -- indexing is 0-based
````

Lists in Haskell are representations of (singly) linked lists:

- operations on lists such as indexing, or computing the length must therefore traverse the list
  - therefore operations like `reverse`, `length` and `!!` are linear in the length of the list
- getting the `head` and `tail` is constant time, as is `:` itself.

## Pattern matching

**Pattern matching** in Haskell is a way of deconstructing values and binding their parts to names based on their shape.  
Instead of writing explicit conditionals (like `if`), functions are defined by matching different patterns of input.

In a function, there can be multiple equations, each handling a different pattern:
```haskell
f 0 = "zero"
f 1 = "one"
f _ = "something else"
```
Here, `0` and `1` are specific patterns, and `_` is a wildcard, which matches anything. 

Pattern matches are tried from top to bottom, and the first match is used.   
Pattern matches should cover all possible cases; otherwise, a runtime error may occur.

### Pattern matching on lists

Pattern matching in lists can be used to recognise specific shapes of lists and extract parts from them.

The patterns must be used in the equations that define the function, not in the type of the function. Pattern matches in the equations don’t change the type of the function - they just say how it should act on particular expressions.

In the example below, `['a', _, _]` is a pattern for a list with exactly three elements. It matches a list where the first element is `'a'` (the other two elements can be anything). 

````haskell
startsWithA :: [Char] -> Bool
startsWithA ['a', _, _] = True
startsWithA _ = False

startsWithA ['a','b','b']
startsWithA ['b','b','b']
````

Note that `String`s are just a list of `Char`s, so pattern matching on lists works on strings:
- `"abc"` is exactly the same as `['a','b','c']`
- double quotes `""` are used for `String`s, and single quotes `''` for `Char`s 

````haskell
startsWithA "abb"  -- "abb" is the same as ['a','b','b']
````

This function only works for lists of length 3; for a list that starts with `'a'` but doesn't have length 3, it will return `False`:

````haskell
startsWithA "a" 
startsWithA "aaaaa"
````

In order to match lists of any length, the `:` operator can be used with the wildcard `_`:

````haskell
startsWithA' :: [Char] -> Bool
startsWithA' ('a': _) = True
startsWithA' _ = False
````

`('a': _)` matches any list of at least length 1 whose first entry is `'a'`.

````haskell
startsWithA' "aaaaa"
````

This method also works to match multiple entries:

````haskell
startsWithAB :: [Char] -> Bool
startsWithAB ('a':'b':_) = True
startsWithAB _ = False

startsWithAB "abcdef"
````

#### Binding variables

As well as just checking for specific literal values, pattern matching can be used to deconstruct data and bind parts of it to variables.

```haskell
sumTwo :: Num a => [a] -> a
sumTwo (x:y:_) = x + y
```

In this function:
- `(x:y:_)` is a pattern that matches any list of length at least 2
- the first element of the list is bound to `x`
- the second element is bound to `y`
- the rest of the list is ignored (using `_`)
- the sum of `x` and `y` is returned

Note that variable names in a single pattern must be unique, though the wildcard can be used multiple times (to ignore the part it matches). 

### Matchable patterns

Pattern matching works by mirroring the structure of the data.  
As a general rule, patterns are constructed to look like the arguments that would be passed to the function.

It can be used to match:
1. Literal patterns
   - matching specific values directly
```haskell
(&&) :: Bool -> Bool -> Bool
True && True = True
False && _ = False
-- matches Boolean literals `True` and `False`
```
2. Constructor patterns
   - matching data built using constructors (e.g., lists)
```haskell
head :: [a] -> a
head (x:_) = x
-- matches any non-empty list
-- binds first element to `x`, ignores rest
```

It is not possible to match arbitrary expressions in patterns; only literals and data constructors are allowed:
```haskell
last :: [a] -> a
last (xs ++ [x]) = x  -- not allowed
-- the ++ operator concatenates two lists
```
Here, `xs ++ [x]` is an expression, not a constructor. Pattern matching only works with the structure of the data, not general computations. 

## List comprehensions

**List comprehensions** in Haskell are a concise way to build lists by describing their elements (similar to set notation in maths).

A list comprehension has the syntax:
```haskell
[ expression | generator, guard ]
```

- `expression` determines what gets added to the output list
- a `generator` has the form `x <- list` and draws values from a list, binding each value to a variable
- `guard`s are optional, and can be used to filter the generated elements

````haskell
doubles = [x * 2 | x <- [1..5]] -- no condition
doubles
````

`x <- [1..5]` draws values of `x` from the list `[1..5]`, and `x * 2` doubles them.

A guard could be used to only include even values of `x`: 

````haskell
evenDoubles = [x * 2 | x <- [1..5], even x] -- filter to only include even x
evenDoubles
````

### Multiple generators

A list comprehension can contain multiple generators, separated by commas. 

Each generator introduces a new variable, and generators are evaluated from left to right. Later generators behave like inner loops and change faster.

````haskell
tuples = [(x,y) | x <- [1,2], y <- [3,4]]
tuples
````

This works like a `for` loop:
```python
for x in [1,2]:
    for y in [3,4]:
        ...
```

It is also possible for a later generator to use variables from earlier ones:

````haskell
tuples' = [(x,y) | x <- [1..3], y <- [x..3]]
tuples'
````

This is equivalent to the `for` loop:
```python
for x in range(1,4):
    for y in range(x,4):
        ...
```

### Guards

Guards are Boolean conditions used to filter results:
- a guard can be any function that returns a Bool
- guards can only use variables that have already been introduced (to their left)

````haskell
evenOdd = [(x,y) | x <- [1..5], even x, y <- [1..5], odd y]
evenOdd
````

This combination of filters creates a list of the tuples `(x,y)`, where `x` is even and `y` is odd, and both are between 1 and 5. 

## Function guards

In functions, the `|` symbol is used to introduce guards. These are conditional branches that are evaluated top-to-bottom until one condition evaluates to `True`.

Take the below Python function:
```python
def sign(x):
    if x > 0:
        return 1
    elif x < 0:
        return -1
    else:
        return 0
```

This can be implemented in Haskell using guards:
```haskell
sign x
    | x > 0 = 1
    | x < 0 = -1
    | otherwise = 0
```

Note that `otherwise` does not have any special meaning; it is conventionally used to represent the fallback case (it is defined as `otherwise = True`). 

## Recursion

Recursion means to define something in terms of itself.

When defining a recursive function in Haskell, the following steps can be followed:

1. define the type
2. enumerate the cases
3. define the simple (base) cases
4. define how other cases reduce to simpler ones
5. (optional) generalise and simplify

For example, suppose we want a function `drop` which drops the first `n` elements from a list. Then, we can follow each of these steps to produce a recursive function:

1. define the type
```haskell
drop :: Int -> [a] -> [a]
```

2. enumerate the cases
```haskell
-- consider all combinations of inputs
-- integer: 0 or non-zero
-- list: empty [] or non-empty (x:xs)
drop 0 [] =
drop 0 (x:xs) =
drop n [] =
drop n (x:xs) =
```

3. define the simple (base) cases
```haskell
-- if n = 0, no elements are dropped, so return the original list
-- if the list is empty, there is nothing to drop, so return []
drop 0 [] = [] 
drop 0 (x:xs) = x:xs
drop n [] = []
drop n (x:xs) =
```

4. define the reduction of other cases to simpler ones
```haskell
-- if n is non-zero and the list is non-empty
-- remove the first element
-- apply drop (n-1) to the rest of the list
drop 0 [] = [] 
drop 0 (x:xs) = x:xs
drop n [] = []
drop n (x:xs) = drop (n-1) xs
```

5. generalise and simplify
```haskell
-- combine cases to make the definition more concise
-- if n = 0, return the list unchanged
-- if the list is empty, always return []
-- otherwise, recurse on the tail
drop 0 xs = xs
drop _ [] = []
drop n (x:xs) = drop (n-1) xs
```

## Equivalence of recursion and iteration

Recursion and iteration are computationally equivalent: any algorithm expressed using one can be transformed into the other.

- the more convenient approach depends on the algorithm and the programming language

To convert from recusion to iteration:

- simple recursion can often be rewritten using loops and variables
- more complex recursion may require an explicit stack to simulate the call stack

To convert from iteration to recursion:

- loop variables become function parameters
- loop is expressed as recursive calls
- use a tail recursive function (where recursive call is the final operation in the function)

## Stacks

A stack is used to manage function calls (including nested and recursive calls).
- a stack is a linear data structure following the Last-In/First-Out (LIFO) principle, where the last element added is the first one removed

Take the function below:
```haskell
length' :: [a] -> Int
length' [] = 0
length' (x:xs) = 1 + length' xs
```

The diagram below illustrates how recursive calls to `length'` are pushed onto the call stack and how results are returned as the stack unwinds when `length' [1,2,3]` is called.

Each entry on the stack uses memory:

- too many entries causes errors (such as stack overflow)
- the size of the stack depends on the language
  - it is typically small in imperative languages and larger in functional ones

## Types of recursive function

Recursive functions can be classified in different ways, which helps when designing efficient implementations.

### Linear recursion

A function is linearly recursive if each call makes at most one recursive call.
```haskell
length' [] = 0
length' (_:xs) = 1 + length' xs
```
The function calls itself repeatedly until it hits the base case.

### Multiple recursion

A function is multiply recursive if each call makes more than one recursive call.
```haskell
fib 0 = 0
fib 1 = 1
fib n = fib (n - 1) + fib (n - 2)
```
This often leads to exponential time complexity due to repeated work.

### Direct recursion

A function is directly recursive if it calls itself.
```haskell
product' [] = 1
product' (x:xs) = x * product' xs
```
Direct recursion can be either linear or multiple, depending on how many recursive calls are made.

### Mutual/indirect recursion
Functions are mutually recursive if they call each other.
```haskell
even' :: Integral a => a -> Bool
even' 0 = True
even' n = odd' (n - 1)

odd' :: Integral a => a -> Bool
odd' 0 = False
odd' n = even' (n - 1)
```

### Tail recursion
A function is tail recursive if the recursive call is the final operation, meaning its result is returned directly with no further computation.

```haskell
drop 0 xs = xs
drop _ [] = []
drop n (x:xs) = drop (n-1) xs
```

Tail-recursive functions are important because:

- they can be optimised to run in constant stack space
- they correspond closely to loops in imperative languages

Tail recursion and iteration are equivalent:

- tail recursion can be translated into loops
- any loop can be rewritten as a tail-recursive function

Consider the following iterative factorial in Python:
```python
def factorial(n):
    res = 1
    for i in range(n, 1, -1):
        res *= i
    return res
```

This can't be written directly in Haskell because variables are immutable (so reassignment like `res *= i` is not allowed). However, it can be turned into a tail recursive function by using a helper recursive function where all the variables in the loop (`res` and `i`) become a parameters:
```haskell
factorial n = loop n 1
    where loop n res
        | n < 0 = undefined
        | n > 1 = loop (n - 1) (res * n)
        | otherwise = res
```

## Lazy evaluation

Haskell is evaluated lazily:

- in **lazy evaluation**, expressions are not evaluated when they are bound to variables; they are evaluated only when their value is required

### Function evaluation strategies

Haskell's basic method of computation is application of functions to arguments. There are two main ways to do this. 

Look at the example function and call below:
```haskell
inc :: Int -> Int 
inc n = n + 1  -- increment input

inc (2 * 3)  -- function call with argument 2*3
```

#### Call-by-value

The arguments are evaluated first, and then the function is applied.
```
inc (2 * 3) 
= inc 6
= 6 + 1
= 7
```

#### Call-by-name
Arguments are substituted into the function body without being evaluated, and evaluation is performed only when required.
```
inc (2 * 3)
= (2 * 3) + 1
= 6 + 1
= 7
```

As long as evaluation terminates (and there are no side effects), the evaluation order does not affect the result.

#### Computation as a graph

A function call and its arguments can be represented as a graph. 

Take the function and call below:
```haskell
mult :: (Int, Int) -> Int
mult (x, y) = x*y

mult (1+2, 3+4)
```

This can be represented by the graph below:

![Computation Graph](computation_graph.png)

Nodes in the graph are either terminals or compounds:

- a terminal node represents an expression that cannot be reduced any further (it is already a final value)
  - `1`, `2`, `3` and `4` are terminals
- a compound node represents an expression that is made up of smaller expressions and can still be evaluated (reduced). These are also called reducible expressions (redexes)
  - `+` and `mult` represent compound nodes

#### Innermost evaluation

In innermost evaluation:

- the graph is evaluated bottom-up
- innermost (deepest) redexes are evaluated first (i.e. those whose arguments are already evaluated values)
- evaluation is typically left-to-right along leaves, though this is implementation-dependent

![Innermost evaluation](innermost.png)

Innermost evaluation corresponds to the call-by-value method for evaluating functions:

- arguments are evaluated before the function is applied
- this is used in most imperative languages

#### Outermost evaluation

In outermost evaluation:

- the graph is evaluated top-down
- outermost redexes are evaluated first
- inner expressions are only evaluated if required by the outer expression
- evaluation is typically left-to-right along children, though this is implementation-dependent

Outermost evaluation corresponds to the call-by-name method for evaluating functions:

- each argument may be evaluated more than once
- Haskell uses a modified version of this (call-by-need) which avoids recomputation by caching results

#### Termination

For expressions that terminate, both innermost and outermost evaluation produce the same result. However, for some expressions involving non-terminating computations, innermost evaluation may fail to terminate, whereas outermost evaluation can still produce a result if the non-terminating part is not required.

Take the example below:
```haskell
inf :: Integer
inf = 1 + inf
fst :: (a, b) -> a
fst (x, _) = x

fst (0, inf)
```
With innermost evaluation, the arguments to `fst` are evaluated first. This requires evaluating `inf`, which does not terminate.  
With outermost evaluation, the `fst` function is applied first. This shows that the second argument is not needed, so `inf` is never evaluated, and the result is `0`.
