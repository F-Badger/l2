# The Entity-Relationship (ER) model

The Entity-Relationship (ER) model is a top-down approach to database design that provides a graphical description of a database. 

It shows:

- **entities**: the important data objects
- **attributes**: the important properties of the entities
- **relationships**: the associations between entities
- **constraints**: rules that apply to entities, attributes, and relationships

## Entities

An **entity type** is a group (or class) of objects with the same properties that are identified as having independent existence.

- Example: `Client`

An **entity** occurrence is a uniquely identifiable instance of an entity type

- Example: a specific client called `James Smith`

An entity type is represented in an ER diagram by a rectangle labelled with the entity type name.

![Entity](er_entity.png)

## Relationships

A **relationship** is a meaningful association between two (or more) entity types.

The **degree** of a relationship is the number of entity types participating in the relationship.

A relationship with degree: 

- 1 is called a unary relationship
- 2 is called a binary relationship
- 3 is called a ternary relationship
- n is called an n-ary relationship

### Binary relationships
A binary relationship is represented by a line connecting the participating entities, labelled with the name of the relationship.

The relationship label may also indicate a reading direction. 

![ER binary relationship representation](er_relationship.png)

### n-ary relationships
To represent an n-ary relationship where $n \geq 3$, a diamond symbol labelled with the relationship name is used to connect the participating entities.

An n-ary relationship does not necessarily have a direction.

![ER n-ary relationship](er_nary_relationship.png)

### Unary relationships
For unary (or recursive) relationships, an entity is related to itself.

This is represented by a relationship line looping from the entity back to itself.

Role names are used to indicate the purpose each entity occurrence plays in the relationship.

![ER unary relationship](er_unary_relationship.png)

### Role names
Role names can also be used when the same entity participates multiple times in a relationship or participates in multiple relationships.

They clarify the purpose of each participating entity in the relationship.

![ER role names](er_role_names.png)

## Attributes

An **attribute** is a descriptive property of an entity or relationship. 

An **attribute domain** is the set of allowable values for an attribute.

Attributes may be:

- simple or composite
  - e.g., `salary` or `address`
- single-valued or multi-valued
  - e.g., `staffID` or `telephoneNums`
- derived
  - e.g., `age` is derived from `dateOfBirth`
- null-valued
  - meaning the attribute currently has no value or the value is unknown/not applicable

### Keys

#### Candidate key
A **candidate key** is a minimal set of attributes whose values uniquely identify an entity occurrence. It cannot contain `null` values.

- minimal means that if any attribute is removed from the set, the remaining attributes no longer uniquely identify the entity occurrence.

A **simple** candidate key consists of a single attribute.

A **composite** candidate key consists of multiple attributes.

#### Primary key
A **primary key** is the candidate key chosen to uniquely identify entity occurrences.

Factors considered when choosing a primary key include:

- number of attributes (preferably smaller)
- attribute length (preferably smaller)
- future certainty of uniqueness

### Attributes on entity diagrams

Attributes are represented on ER diagrams by dividing the entity rectangle into two sections:

- the upper section contains the entity name
- the lower section contains the list of attributes

The primary key is usually:
- underlined, or
- labelled with the tag `{PK}`

![ER attributes](er_attributes.png)

### Attributes on relationships

Relationships can also have attributes.

These describe properties of the relationship itself rather than properties of the participating entities.

To represent these, a labelled diagram with two parts is again used:

- the upper section is empty
- the lower section contains the list of relationship attributes

The rectangle is connected by a dashed line with the relationship.

![ER relationship attribute](er_relationship_attributes.png)

## Multiplicity

The multiplicity (or cardinality) of a relationship specifies the number of entity occurrences that can participate in the relationship.

Relationships can be:

- one-to-one (1:1)
  - e.g., one `Staff` manages one `Branch`
- one-to-many (1:*)
  - e.g., one `MathTeacher` teaches many `Student`s
- many-to-many (\*:\*)
  - e.g., Many `Newspaper`s advertise many `PropertyForRent`
 
In an ER diagram, multiplicity constraints are shown on the relationship lines. They specify the minimum and maximum number of relationship occurrences in which an entity occurrence may participate.

![ER Multiplicity](er_multiplicity.png)

### Multiplicity of n-ary relationships

For an n-ary relationship, multiplicity specifies the number of entity occurrences that may participate in the relationship when occurrences of the other $n - 1$ participating entity types are fixed.

![ER n-ary multiplicity](er_nary_multiplicity.png)

In this example:

- for fixed `StaffNo` and `BranchNo`, there may be zero or more `ClientNo` values
  - each `Staff` member at a `Branch` may register 0 or more `Client`s
- for fixed `StaffNo` and `ClientNo`, there is exactly one `BranchNo` value
  - each `Staff` - `Client` pair is associated with exactly one `Branch`
- for fixed `ClientNo` and `BranchNo`, there is exactly one `StaffNo` value
  - each `Client` is registered at their `Branch` by exactly one `Staff` member

Therefore, the multiplicity of `Registers` is:

- with respect to `Staff` and `Branch`: `0..*` (many)
- with respect to `Staff` and `Client`: `1..1` (one)
- with respect to `Client` and `Branch`: `1..1` (one)

## Problems with the ER model

The basic concepts of the ER model (entities, attributes and relationships) are generally adequate for modelling simpler databases.

However, the ER model becomes limiting when modelling more complex database applications that require richer semantics and more advanced structural constraints.

For this reason, the Enhanced Entity-Relationship (EER) model was developed as an extension of the ER model.

## The Enhanced Entity-Relationship (EER) model

The Enhanced Entity-Relationship (EER) model extends the ER model by introducing additional concepts that allow more expressive and intuitive database design.

### Superclasses and subclasses

The EER model introduces the concepts of superclasses and subclasses:

- A **subclass** is a subset of an entity type whose occurrences share additional properties and require separate representation.
- A **superclass** is an entity type that has one or more distinct subclasses.

For example, the superclass `Staff` could have subclasses `Manager` and `Secretary`. 

Each member of a subclass is also a member of the superclass.

#### Inheritance

All attributes and relationships of the superclass are inherited by its subclasses.

A subclass may also have additional attributes and relationships that are not present in the superclass.

#### Type hierarchy

A type hierarchy in the EER model is the organisation of entity types into a superclass-subclass structure in which subclasses inherit attributes and relationships from a more general superclass.

It can also be known as:

- Specialisation hierarchy
  - e.g., `Manager` is specialisation of `Staff`
- `Generalisation` hierarchy
  - e.g., `Staff` is generalisation of `Manager`
- IS-A hierarchy
  - e.g., `Manager` IS-A (member of) `Staff`

### Advantages

The main advantages of the EER are that:

- similar concepts do not need to be repeatedly described, as common properties can be stored in a superclass and inherited by subclasses
- relationships can involve subclasses without involving the entire superclass, allowing more precise modelling
- it adds additional semantic information to the database design, making it closer to real-world meaning
  - e.g.,`Manager` IS-A member of `Staff`, `Van` IS-A type of `Vehicle`

### Specialisation

**Specialisation** is one approach to producing an EER model:

- it is a top-down process of maximising the differences between entity occurrences by identifying their distinguishing characteristics
- starting from one or more superclasses, it leads to the identification of more specific subclasses

### Generalisation

**Generalisation** is the opposite approach:

- it is a bottom-up process of minimising differences between entity occurrences by identifying their common characteristics
- starting from subclasses, it leads to the formation of a more general superclass

### Constraints

#### Participation constraint

The participation constraint determines whether every entity in a superclass must belong to a subclass (total participation), or whether some entities may not belong to any subclass (partial participation).

It can be:

- mandatory (total specialisation): every superclass entity must belong to at least one subclass
- optional (partial specialisation): some superclass entities may not belong to any subclass

#### Disjoint constraint

The disjoint constraint determines whether a superclass entity can belong to more than one subclass.

It only applies when there are at least two subclasses.

It can be:

- disjoint (or): an entity can belong to only one subclass
- overlapping (and): an entity can belong to more than one subclass

### EER diagrams

In EER diagrams, specialisation/generalisation relationships are shown using a triangle (IS-A) symbol connecting the superclass and its subclasses.

- The triangle is labelled (or implied) as an IS-A relationship
- Connection lines from the subclasses meet at the triangle
- The triangle points towards the superclass
- Constraints are shown as a label near/below the triangle

# Semi-structured data - XML

## Types of data organisation

Data can be classified into three categories:

- structured data
- semi-structured data
- unstructured data

### Structured data 

Structured data is data that is represented in a strict format, defined by the schema.

- for example, in the relational data model (with tuples, tables and attributes)

The database management system (DBMS) enforces the schema to ensure that the data conforms to it, including:

- structure (tables, attributes, domains)
- integrity and referential constraints (primary keys and foreign keys)

### Semi-structured data

Semi-structured data is data that:

- does not conform to a rigid, fixed schema (like relational tables), but
- still contains some form of structure within the data itself

It may be:

- irregular or incomplete
- flexible, with structure that can change over time
- heterogeneous (different data items may have different attributes)

This data may have some structure, but:

- not all the parts have the same fixed structure
- each data object may have different attributes that are not known in advance

Semi-structured data is often described as self-describing or schema-less data:

- the structure information is embedded within the data itself
- data and schema information are mixed together

In some cases, there may exist a separate schema, such as a Document Type Definition (DTD) or XML Schema Definition (XSD). These provide optional constraints but are not as rigid as relational schemas.

### Unstructured data

Unstructured data has no predefined data model or strict structure. It does not follow a schema that can be easily represented in tables or fields

For example, a plain text document or image. 

## eXtensible Markup Language (XML)

`eXtensible Markup Language (XML)` is one of the most commonly used languages for representing semi-structured data:

- it is commonly used for storing, structuring and exchanging data, especially on the web

XML describes both the structure and content (semantics) of data.

## Tags and terminology

XML uses tags (like HTML), which are enclosed using `<>` symbols.  

Shown below is an example of some XML:

```xml
<bibliography>
    <book>
        <title> Foundations of Databases </title>
        <author> Abiteboul </author>
        <author> Hull </author>
        <author> Vianu </author>
        <publisher> Addison Wesley </publisher>
        <year> 1995 </year>
    </book>
    …
</bibliography>
```

This example can be used to illustrate some key XML terminology:

- tags: custom labels that describe what the data represents
  - for example, `book`, `title`, `author`
  - a start tag is enclosed in `<>`, for example `<book>`
  - an end tag is enclosed in `</>`, for example `</book>`
- elements: sections enclosed in start and end tags
  - for example, `<book> ... </book>`
  - elements may contain text, attributes or other nested elements
  - elements can be nested
  - elements can also be empty
    - for example, `<author></author>` or `<author/>`
- root element: the single top-level element that contains all other elements
  - in this example, `bibliography`

An XML document must contain exactly one root element.

Note that XML is case-sensitive. 

## XML data as a graph

XML data can be represented as a directed tree structure. 

![XML data as a graph](xml_graph.png)

In the tree:

- the document node is the root of the tree
- the root element of the XML is the child of the document node
- internal nodes are element nodes
- leaf nodes are text nodes (the actual data content)
- XML is order-sensitive: if two sibling elements are swapped, the structure and traversal order of the tree also changes

Query languages for XML navigate and query this tree-based representation. 

## Advantages of XML

There are many advantages of using XML to represent data:

- simplicity: it has a relatively simple, human-readable syntax
- extensibility: users can define their own tags for specific application needs
- platform- and vendor-independent: XML is widely supported and uses Unicode, enabling international character support
- separation of content and presentation: XML stores data independently of how it is displayed; presentation can be handled separately, allowing data to be reused in different formats

## XML attributes

Attributes are a name-value pair with descriptive or identifying information about an element:

- they are placed inside the start tag of the element
- the attribute value is enclosed in quotes `""`
- e.g., `<book id="b1">`
- attributes are associated with element nodes in the tree representation

### Attributes vs elements

Attributes and elements are similar, but there are some important differences:

- an attribute name can appear only once within an element's start tag
- subelements can appear multiple times
  - subelements may contain the same attribute names
- attributes are unordered (subelements are ordered)
- attributes cannot contain nested structure (subelements can)
- attributes are generally less flexible and less extensible than elements
- attributes are typically used to store metadata (data about data)

### ID and IDREF

Some XML attributes can be declared as specific types in a DTD:

- `ID`: used as a unique identifier for an element
- `IDREF`: used as a reference to an element with a specific `ID`
  - if an attribute is declared as `IDREF`, its value must match an existing `ID`
- `IDREFS`: an extension of `IDREF` that allows references to multiple elements
  - the value is a whitespace-separated list of `ID` values

The constraints associated with these types are defined and validated by the DTD (or schema).

`ID` and `IDREF` are similar to primary keys and foreign keys in the relational model.
 
When `IDREF` or `IDREFS` attributes are used, the XML structure is no longer purely a tree; it becomes a directed graph, which may contain cycles.

The example below shows some XML that uses `ID` and `IDREF` attributes
```xml
<person id="o555">
    <name>Jane</name>
</person>
<person id="o456" children-idrefs="o123 o555">
    <name>Mary</name>
</person>
<person id="o123" mother-idref="o456">
    <name>John</name>
</person>
```

## More XML syntax

### Comments

Comments in XML are enclosed in `<!--` and `-->` tags:

- they can contain any data except the string `--`
- they are ignored by XML parsers

### CDATA

CDATA sections are used to include unparsed character data:

- can contain any text
- text inside CDATA is not treated as XML markup
- special characters (e.g. `<`, `&`) do not need to be escaped

### Entity reference

Entity references are used to represent reserved or special characters in XML.

- they begin with `&` and end with `;`
- during parsing, they are replaced with their corresponding character
  - e.g., `&amp;` will be replaced with `&`

## XML validation

Although XML is self-describing in terms of structure, additional constraints can be imposed to ensure consistency and correctness.

These constraints may define:

- permissible element names
- which elements can appear together
- how elements can be nested
- what attributes are allowed for each element type

### Document Type Definition (DTD)

A Document Type Definition (DTD) is a formal set of rules that defines the structure of an XML document:

- specifies allowed elements and attributes
- defines how elements can be nested
- ensures consistent structure between XML documents
- appropriate for specific applications

The general structure of a DTD is:
```dtd
<!DOCTYPE root_name [
    <!ELEMENT elem_name (subElem1, subElem2, …)>
    ... more element declarations ...
]>
```

An element declaration defines the content model of an element:

- the element name (tag)
- the structure of its content (e.g. sequence, choice, repetition)
- e.g., `<!ELEMENT book (title, author+)>`
  - this means: a `book` contains a `title` followed by one or more `author` elements

`#PCDATA` stands for Parsed Character Data:

- it represents text content inside an element
- it is not a sub-element
- e.g., `<!ELEMENT title (#PCDATA)>`

#### Using the DTD

To use the DTD, it can either be embedded directly within the XML file:
```xml
<?xml version="1.0" standalone="no"?>
<!DOCTYPE root_name [
    <!ELEMENT elem_name (subElem1, subElem2, …)>
    ... more element declarations ...
]>
<root_name>
        ...
</root_name>
```

Or, the DTD is stored in a separate file and referenced using `SYSTEM`:
```xml
<?xml version="1.0" standalone="no"?>
<!DOCTYPE root_name SYSTEM "filename.dtd">
<root_name>
        ...
</root_name>
```

#### Sub-elements in a DTD

Sub-elements are defined in the DTD content model and correspond to element names used in XML:

- sub-elements appear nested within an element in the XML document
- the order of sub-elements in XML must match the order specified in the DTD when using sequences
- commas define a sequence (ordered list) of sub-elements
- if commas are omitted, then they can appear in an arbitrary order
- the `|` symbol defines a choice between sub-elements

The multiplicity of a sub-element is specified by:

- `*`: zero or more
- `+`: one or more
- `?`: = zero or one

Furthermore, `|` can be used to mean 'or', providing a choice between sub-element names. 

#### Attributes in a DTD

Attribute list declarations (`<!ATTLIST>`) define:

- which elements may have which attributes
- what types or values those attributes can take
- whether they are of type `ID`, `IDREF`, or `IDREFS`

DTD provides keywords to control whether attributes must appear:

- `#REQUIRED`: the attribute must always be provided
- `#IMPLIED`: the attribute is optional
- `"value"`: specifies a default value if the attribute is omitted

For example:
```dtd
<!ELEMENT book (title, author)>

<!ATTLIST book
    id        ID       #REQUIRED
    genre     CDATA    #IMPLIED
    language  CDATA    "English"
>
```

In this example:

- `id` must always be provided
- `genre` is optional
- `language` defaults to `"English"` if omitted

### XML schema

An XML Schema (XSD) is a more powerful and flexible alternative to DTD:

- supports complex structural constraints
- includes built-in data types
- written in XML syntax
- allows more precise validation rules than DTD

### Well-formed XML

The first type of XML validation ensures that XML is well-formed:

- there is exactly one root element
- all elements are properly nested with matching start and end tags
- there is a valid XML declaration (optional)
  - e.g., `<?xml version="1.0" encoding="UTF-8" standalone="yes"?>`
  - `standalone="yes"` indicates that the document does not depend on external declarations (such as external DTDs or entities)

### Type-valid XML

The second type of XML validation ensures that XML is type-valid:

- the document must be well-formed
- elements and attributes must follow rules defined in a DTD

### Schema-valid XML

The third type of XML validation ensures that XML is schema-valid:

- the document must be well-formed
- it must conform to an XML Schema (XSD)

# XPath

**XML Path Language** (**XPath**) is a language used to navigate and select parts of an XML document. 

It works by describing paths through the XML tree structure.

XPath:

- is a declarative query language for XML
- provides a simple syntax for addressing parts of an XML document
- treats an XML document as an ordered tree of nodes
- uses location paths composed of a series of steps joined with `/` (similar to directory paths)

## XPath queries

For this section, the XML extract below will be used as an example:

```xml
<bib>
    <book>
        <publisher>Addison-Wesley</publisher>
        <author>Serge Abiteboul</author>
        <author>
            <firstName>Rick</firstName>
            <lastName>Hull</lastName>
        </author>
        <author>Victor Vianu</author>
        <title>Foundations of Databases</title>
        <year>1995</year>
    </book>

    <book price="55">
        <publisher>Freeman</publisher>
        <author>Jeffrey D. Ullman</author>
        <title>Principles of Database and Knowledge Base Systems</title>
        <year>1998</year>
    </book>
</bib>
```

### Paths

An XPath expression selects the nodes matching a path through the XML tree.

- starting with `/` denotes an absolute path starting from the document root
- each `/` separates a navigation step

#### Example 1
```xpath
/bib/book/year
```
selects
```xml
<year>1995</year>
<year>1998</year>
```

#### Example 2
```xpath
/bib/paper/year
```
returns an empty result because there are no `paper` elements in the document. 

### Descendant search

Two forward slashes `//` can be used in an XPath expression to select matching descendant nodes at any depth in the XML tree

- it searches recursively through descendants of the current context node

#### Example 1
```xpath
//author
```
selects
```xml
<author>Serge Abiteboul</author>
<author><firstName>Rick</firstName>
        <lastName>Hull</lastName>
</author>
<author>Victor Vianu</author>
<author>Jeffrey D. Ullman</author>
```

#### Example 2
```xpath
/bib/book//firstName
```
selects
```xml
<firstName>Rick</firstName>
```

### Wildcards

The wildcard symbol `*` matches any element node at a particular step in an XPath expression.

- it matches element nodes only (not text nodes or attributes)

#### Example 1
```xpath
/bib/book/author/*
```
selects
```xml
<firstName>Rick</firstName>
<lastName>Hull</lastName>
```
because only one author element contains child elements.

#### Example 2
```xpath
//book/*
```
selects
```xml
<publisher>Addison-Wesley</publisher>
<author>Serge Abiteboul</author>
<author>
    <firstName>Rick</firstName>
    <lastName>Hull</lastName>
</author>
<author>Victor Vianu</author>
<title>Foundations of Databases</title>
<year>1995</year>

<publisher>Freeman</publisher>
<author>Jeffrey D. Ullman</author>
<title>Principles of Database and Knowledge Base Systems</title>
<year>1998</year>
```

### Attribute nodes

The `@` symbol is used to select attributes:

- `@attribute` selects the attribute node named `attribute`
- `@*` selects all attributes of an element

#### Example
```xpath
/bib/book/@price
```
selects the `price` attribute value:
```xml
55
```

### Predicates

A predicate can be added after a node in an XPath expression:

- a predicate is an expression that evaluates to true or false
- it is written in square brackets `[]` after a node
- only nodes that satisfy the predicate are selected
- predicates often imply an existence test for an element or attribute

#### Example 1
```xpath
//book[@price="55"]
```
selects all `book` elements with a `price` attribute equal to `"55"`:
```xml
<book price="55">
    <publisher>Freeman</publisher>
    <author>Jeffrey D. Ullman</author>
    <title>Principles of Database and Knowledge Base Systems</title>
    <year>1998</year>
</book>
```

#### Example 2
```xpath
/bib/book/author[firstName]
```
selects all `author` elements that contain at least one `firstName` child element:
```xml
<author>
    <firstName>Rick</firstName>
    <lastName>Hull</lastName>
</author>
```

### Multiple predicates

A query can contain multiple predicates. Each predicate is written in square brackets [].

- predicates are applied from left to right
- each predicate filters the result of the previous step

If predicates are independent (i.e. they do not depend on position or different contexts), their order does not affect the final result. 

#### Example
```xpath
/bib//book[@price="55"][.//firstName | publisher = "Freeman"]
```
selects all `book` elements that:

- have a `price` attribute equal to `"55"`
- and either:
  - contain a `firstName` element anywhere inside them, or
  - have a `publisher` child element with value `Freeman`
```xml
<book price="55">
    <publisher>Freeman</publisher>
    <author>Jeffrey D. Ullman</author>
    <title>Principles of Database and Knowledge Base Systems</title>
    <year>1998</year>
</book>
```
Since these two predicates are independent, they can be reversed without changing what is selected.

### Axes

The general form of an XPath query is 
```xpath
/step-1/step-2/step-3/.../step-n
```

Each of these steps consists of:

- a basis, and
- a list of predicates (optional)

A basis consists of:

- an axis: the direction in which the navigation proceeds from the current node
- a node test: the type or name of node to navigate to

The default axis is:
```xpath
child::
```

So expressions like:
```xpath
/bib/book
```
are equivalent to:
```xpath
child::bib/child::book
```

There are many other axes, such as:

- `parent::` - selects the parent of the current node
- `attribute::` - selects attributes of the current node
- `self::` - selects the current node itself
- `child::` - selects child elements (default axis)
- `descendant::` - selects all descendants of the current node
- `descendant-or-self::` - selects the current node and all descendants
- `ancestor::` - selects all ancestors of the current node
- `ancestor-or-self::` - selects ancestors and the current node
- `following-sibling::` - nodes with the same parent that come after the current node
- `preceding-sibling::` - nodes with the same parent that come before the current node
- `following::` - all nodes in document order after the current node
- `preceding::` - all nodes in document order before the current node

#### Syntax

An axis for a tag (or attribute) is selected by prefixing it with an axis name and a double colon `::`.

##### Example 1
```xpath
/bib/book/@price
```
is equivalent to:
```xpath
/child::bib/child::book/attribute::price
```

##### Example 2
```xpath
//book/*
```
is equivalent to:
```xpath
/descendant-or-self::book/child::*
```

### Functions

Two key XPath functions are:

- `text()`: selects the text node children of an element
- `name()`: returns the name of the current node as a string

#### Example 1
```xpath
/bib/book/author/text()
```
selects 
```xml
Serge Abiteboul
Victor Vianu
Jeffrey D. Ullman
```

#### Example 2
```xpath
/bib//*[name() = "book"]
```
is equivalent to `/bib//book` and selects
```xml
<book>
    <publisher>Addison-Wesley</publisher> 
    ...
</book>
<book price="55">
    <publisher>Freeman</publisher>
    ...
</book>
```

### Positional predicates

The position() function returns the position of a node within the current node set being evaluated.

It is used to select nodes based on their position relative to sibling nodes.

#### Example
```xpath
//author[position() = 2]
```
selects the second `author` child element of each parent node:
```xml
<author> <firstName>Rick</firstName>
         <lastName>Hull</lastName>
</author>
```

The same result can be written more concisely as:
```xpath
//author[2]
```

# XQuery

XQuery is a powerful query and transformation language for XML.

It is often described as the SQL equivalent for XML data.

It:

- uses XPath to navigate XML data
- selects and filters nodes
- combines and transforms data
- constructs new XML or other output formats

## Basic query form

The basic form of an XQuery query is a FLWR expression:
```xquery
FOR ...
LET ...
WHERE ...
RETURN ...
```

The purpose of each clause is as follows:

- `FOR`: iterates over a sequence of nodes
- `LET`: binds a sequence to a variable
- `WHERE`: filters nodes based on a condition
- `RETURN`: constructs the output

### `FOR-WHERE-RETURN`

A `FOR-WHERE-RETURN` expression iterates over nodes, filters them, and returns results.

#### Example

```xquery
FOR $x IN doc("bib.xml")/bib/book
WHERE $x/year > 1995
RETURN { $x/title }
```

In this query:

- `FOR $x IN doc("bib.xml")/bib/book` iterates over each `book` element selected by the XPath expression `doc("bib.xml")/bib/book`, and binds it to the variable `$x`
- `WHERE $x/year > 1995` filters books whose `year` value is greater than 1995
- `RETURN { $x/title }` returns the `title` element of each selected book

Continuing with the [example XML from earlier](#XPath-queries), the result would be:
```xml
<title>Database System Implementation</title>
<title>Introduction to Databases</title>
```

This query could be written equivalently in XPath as
```xpath
doc("bib.xml")/bib/book[year > 1995]/title
```

### `FOR` vs `LET`

`FOR` binds a variable to each item in a sequence one at a time:

- it iterates over a sequence of nodes
- the variable is rebound for each item

For example
```xquery
FOR $x IN doc("bib.xml")/bib/book
```
means `$x` takes each `book` element one by one during iteration. 

`LET` binds a variable to an entire sequence at once:

- no iteration occurs
- the variable stores the full collection

For example
```xquery
LET $x := doc("bib.xml")/bib/book
```
means `$x` contains the full sequence of all `book` elements.|

### Collections

A collection is a sequence of items bound to a variable using `LET`.

- sequences are ordered by default
- sequences may contain duplicate values or nodes

The function:
```xquery
distinct-values(...)
```
is used to remove duplicate atomic values from a sequence.

For example,
```xquery
distinct-values(/bib/book/author/text())
```
returns a sequence of unique author names.

### `WHERE`

The `WHERE` clause contains one or more conditions used to filter results produced by the `FOR` and `LET` clauses.

For variables bound by a `FOR` clause:

- the variable is evaluated one item at a time
- predicates are typically applied to individual elements or values
- commonly used with scalar conditions
  - e.g., `$x/salary > 10000`
 
For variables bound by a `LET` clause:

- the variable usually contains an entire sequence
- predicates are often applied to the sequence as a whole
- commonly used with aggregate or list-oriented functions
  - e.g., `avg($S/SALARY) > 10000`

Common aggregate functions include:

- `count()`: returns the number of items in a sequence
- `avg()`: returns the average of numeric values in a sequence
- `min()`, `max()`: return the minimum and maximum values respectively
- `sum()`: returns the sum of numeric values in a sequence
- `string-length()`: returns the length of a string
- `empty()`: checks whether a sequence is empty
- `exists()`: checks whether a sequence contains at least one item
 
In XPath/XQuery comparison operations, automatic type processing is performed:

- atomization: converts nodes (such as elements) into their atomic values
- casting: converts atomic values into an appropriate type for comparison (e.g. integer, string)

### Joins

XQuery supports joins between sequences of XML nodes, similar to joins in SQL.

For example:
```xquery
FOR $S IN doc("staff_list.xml")//Staff,
    $B IN doc("branch.xml")//Branch
WHERE $S/branchNo = $B/branchNo
RETURN
    <staff-branch>
        {$S}
        {$B/Address}
    </staff-branch>
```
- the query iterates over all Staff and Branch elements
- the `WHERE` clause matches staff members to branches with the same `branchNo`
- the `RETURN` clause constructs a new XML element combining the matched data

### Ordering

The `ORDER BY` clause is used to control the order in which query results are returned:

- it sorts the results of a `FLWR` expression
- sorting is usually based on element or attribute values
- ascending order is the default
- descending can be specified explicitly

For example,
```xquery
FOR $b IN doc("bib.xml")/bib/book
ORDER BY $b/year
RETURN $b/title
```
returns book titles ordered by year.

### `IF-THEN-ELSE`

XQuery supports conditional branching using `IF-THEN-ELSE` expressions:

```xquery
FOR $h IN doc("bibliography.xml")//papers
ORDER BY $h/title
RETURN
    <holding>
        {
            IF ($h/@type = "Journal")
            THEN $h/editor
            ELSE $h/author[1]
        }
    </holding>
```
In this example:

- if the `type` attribute of the current paper is `"Journal"`, then the `editor` element is returned
- otherwise, the first `author` child element is returned

### Existential qualifiers: `SOME`

The `SOME` keyword is used to test whether at least one item in a sequence satisfies a condition.
```xquery
FOR $b IN doc("bibliography.xml")//book
WHERE SOME $p IN $b//chapter SATISFIES
            (
                contains($p, "sailing") 
                AND $p/length < 40
            )
RETURN {$b/title}
```
This query returns the `title` of each `book` where at least one `chapter`:

- contains the word `"sailing"`
- and has a `length` value less than 40

### Universal qualifiers: `EVERY`

The `EVERY` keyword is used to test whether all items in a sequence satisfy a condition.
```xquery
FOR $b IN doc("bibliography.xml")//book
WHERE EVERY $p IN $b//chapter SATISFIES
        (
            contains($p, "sailing") 
            AND $p/length < 40
        )
RETURN {$b/title}
```
This query returns the `title` of each `book` where every `chapter`:

- contains the word `"sailing"`
- and has a `length` value less than 40

# Transactions

## Database Management Systems

A Database Management System (DBMS) is software used to create, store, manipulate, and manage databases efficiently.

A DBMS must ensure that a database:

- is reliable
  - operations behave correctly and as expected
- remains consistent
  - the database does not enter an invalid or contradictory state

To achieve this, the DBMS must be able to handle:

- software and hardware failures
- multiple users accessing the database concurrently

These issues are managed using:

- database recovery
  - the process of restoring the database to a correct and consistent state after a failure
- concurrency control protocols
  - mechanisms that prevent simultaneous transactions from interfering with one another
 
To support recovery and concurrency control, DBMSs use the concept of transactions.

## Transactions

A **transaction** is a sequence of one or more database operations performed as a single logical unit of work by a user or program

Transactions may:

- read data
- modify data
- insert data
- delete data

A transaction usually represents a single real-world task, such as transferring money between bank accounts. 

At the end of a transaction:

- the database must be in a consistent state
  - all integrity constraints (such as primary key and foreign key constraints) are satisfied

During execution of a transaction:

- the database may temporarily be in an inconsistent state
  - some constraints may be temporarily violated until the transaction completes
 
A transaction can have two outcomes:

- committed
  - the transaction completed successfully, so all changes are permanently saved to the database
- rolled back (or aborted)
  - the transaction failed or was cancelled, so all changes made by the transaction are undone

### Properties of transactions

All transactions must satisfy the ACID properties:

- atomicity
- consistency
- isolation
- durability

#### Atomicity

Atomicity means that a transaction is performed entirely or not performed at all.

If any part of the transaction fails, the entire transaction is rolled back and the database is restored to its previous state.

The responsibility for atomicity lies with the recovery subsystem of the DBMS.

#### Consistency

Consistency means that a transaction must transform the database from one consistent state to another consistent state.

After a transaction completes, all integrity constraints (such as primary key and foreign key constraints) must still hold.

Responsibility for consistency lies with both the DBMS and the application developers.

- for example, the DBMS can enforce integrity constraints, but incorrect transaction logic written by a programmer may still produce invalid data

#### Isolation

Isolation means that concurrently executing transactions do not interfere with one another.

- the intermediate (partial) effects of an incomplete transaction should not be visible to other transactions

From the perspective of each transaction, it should appear as though it is executing alone on the database.

Responsibility for isolation lies with the concurrency control subsystem of the DBMS.

#### Durability

Durability means that once a transaction has been committed, its changes are permanently stored and will not be lost due to failures such as software crashes or power loss.

Responsibility for durability lies with the recovery subsystem of the DBMS.

## Concurrency control

Concurrency control is the process of managing simultaneous operations on a database so that transactions do not interfere with one another.

Two transactions may each be correct when executed independently, but if they execute concurrently, they may produce an inconsistent database state.

Common problems caused by concurrent transactions include:

- lost update problem
- uncommitted dependency problem
- inconsistent analysis problem

### Lost update problem

The lost update problem occurs when two transactions both update the same data item and one update overwrites the other.

As a result, one transaction's update is effectively lost, even though both transactions completed successfully.

![Lost update problem](lost_update.png)

### Uncommitted dependency problem

The uncommitted dependency problem occurs when a transaction is allowed to read data that has been modified by another transaction that has not yet committed.

If the other transaction is later rolled back, the first transaction will have used invalid data.

This problem is also known as a dirty read.

![Uncommitted dependency problem](uncommitted_dependency.png)

### Inconsistent analysis problem

The inconsistent analysis problem occurs when a transaction reads a set of data while another transaction is updating part of that data.

As a result, the transaction may see some old values and some new values, producing inconsistent results.

This commonly occurs when one transaction is performing an aggregate operation such as SUM or COUNT while another transaction updates the same data.

![Inconsistent analysis problem](inconsistent_analysis.png)

## Schedules

An obvious solution to concurrency problems is to allow only one transaction to execute at a time.

- i.e. one transaction completes (commits or rolls back) before the next transaction begins

However, this approach greatly reduces concurrency and system performance.

To maximise concurrency while still maintaining database consistency, the DBMS uses concurrency control mechanisms.

One important concept used in concurrency control is a schedule.

One such mechanism is schedules:

- a **schedule** is an ordering of the operations from a set of concurrent transactions $T_1, T_2, \ldots, T_n$ such that the order of operations within each individual transaction is preserved.
- a **serial schedule** is a schedule in which the operations of different transactions are not interleaved
  - one transaction executes completely before another begins
- a **non-serial schedule** is a schedule in which operations from different transactions are interleaved

### Serialisable schedules

Any serial schedule leaves the database in a consistent state, although different serial schedules may produce different valid database states.

A non-serial schedule is serialisable if it produces the same final database state as some serial schedule involving the same transactions.

In other words, a serialisable schedule behaves equivalently to a serial execution of those transactions.

#### Operation conflicts

Two operations are not in conflict if:

- they are both read operations on the same data item
- they operate on different data items

Two operations are in conflict if they belong to different transactions, operate on the same data item, and at least one of them is a write operation.

In serialisability, only the ordering of conflicting operations matters.

- non-conflicting operations may be reordered without affecting the correctness of the schedule

#### Conflict serialisability

Conflict serialisability ensures that a concurrent schedule produces the same effect as some serial execution of the same transactions.

To determine whether a non-serial schedule is conflict serialisable, a precedence graph (or serialisation graph) can be constructed.

A precedence graph is a directed graph $G = (V,E)$, where:

- each node $T_i \in V$ represents a transaction
- a directed edge $T_i \to T_j$ exists if:
  - an operation in $T_i$ conflicts with an operation in $T_j$, and
  - the operation in $T_i$ occurs before the conflicting operation in $T_j$ in the schedule

In the precedence graph, an edge $T_i \to T_j$ means that $T_i$ must appear before $T_j$ in any equivalent serial schedule.

A schedule is conflict serialisable if and only if its precedence graph contains no directed cycles.

#### View serialisability

View serialisability is another form of serialisability based on whether transactions observe the same values.

Two schedules $S_1$ and $S_2$ are view equivalent if all of the following conditions hold:

- for each data item $x$, if transaction $T_i$ reads the initial value of $x$ in $S_1$, then $T_i$ also reads the initial value of $x$ in $S_2$
- for each read operation $r_i(x)$ in $S_1$, if transaction $T_i$ reads the value of $x$ written by transaction $T_j$, then in $S_2$, $T_i$ must also read the value of $x$ written by $T_j$
- for each data item $x$, if transaction $T_i$ performs the final write on $x$ in $S_1$, then $T_i$ must also perform the final write on $x$ in $S_2$

In other words, two schedules are view equivalent if:

- the same transactions read the same values, and
- the same transaction performs the final write on each data item

A non-serial schedule is view serialisable if it is view equivalent to some serial schedule. 

Every conflict serialisable schedule is also view serialisable. However, the converse is not always true.

## Concurrency control approaches

There are two main approaches used to ensure consistency when transactions execute concurrently:

- conservative (pessimistic) methods
  - assume that conflicts between transactions are likely
  - actively prevent conflicts from occurring
  - delay, block, or restart transactions when conflicts are detected
- optimistic methods
  - assume that conflicts between transactions are rare
  - allow transactions to execute freely without immediate restrictions
  - check for conflicts only before commit (during validation)

### Locking 

Locking is a conservative (pessimistic) approach to concurrency control.

A **lock** is a mechanism used by the DBMS to control concurrent access to data items and prevent transactions from interfering with one another.

A transaction T can keep two types of locks:

- shared lock (or read lock):
  - T is allowed to read the data item
  - other transactions may also acquire shared locks on the same item and read it
  - no transaction may write to the item while shared locks exist
- exclusive lock (or write lock):
  - T is allowed to read and write the data item
  - no other transaction may read or write the item while the exclusive lock is held

Locks are used as follows:

- a transaction requests a lock from the DBMS
- if the lock is compatible with existing locks, the DBMS grants it
- otherwise, the transaction is blocked (waits) until the lock can be granted
- when the transaction no longer needs the data item, it releases the lock

#### Rules for locks

When a transaction needs to access a data item, it requests either:
- a shared lock for read-only access, or
- an exclusive lock for read and write access

Then:
- if no other transaction holds an incompatible lock on the item, the requested lock is granted
- otherwise, the DBMS checks whether the requested lock is compatible with the existing lock(s):
  - if the item currently has only shared locks and another shared lock is requested, the lock is granted
  - otherwise, the requesting transaction must wait until the conflicting lock is released

A transaction holds a lock until it explicitly releases it:

- during execution, once the item is no longer needed, or
- when the transaction terminates (commits or aborts)

Exclusive locks prevent other transactions from reading or writing the locked item while the transaction is updating it.

This helps ensure the Isolation property of transactions.

To improve concurrency, some systems may allow locks to be:

- upgraded from a shared lock to an exclusive lock, or
- downgraded from an exclusive lock to a shared lock

#### Two-phase locking

Locking alone does not guarantee serialisability. 

To guarantee conflict serialisability, an additional locking protocol called two-phase locking (2PL) is used.

In two-phase locking, each transaction has two distinct phases:

- growing phase
  - the transaction may acquire locks
  - no locks may be released
- shrinking phase
  - the transaction may release locks
  - no new locks may be acquired

Therefore, once a transaction releases its first lock, it cannot obtain any additional locks.
 
If every transaction in a schedule follows the two-phase locking protocol, then the schedule is guaranteed to be conflict serialisable.

#### Cascading rollback problem

Although 2PL guarantees conflict serialisability, problems can still occur if locks are released before a transaction commits.

This can lead to the cascading rollback problem. 

A cascading rollback occurs when one transaction reads or uses data written by another uncommitted transaction that later aborts.

As a result, the dependent transaction must also be rolled back, potentially causing a chain of rollbacks.

For example:

- $T_1$ obtains an exclusive lock on data item x, updates x, and releases the lock before committing
- $T_2$ then accesses and updates the modified value of x
- $T_3, T_4, \ldots, T_n$ continue using values derived from previous uncommitted transactions
- later, $T_1$ aborts and its update to x is undone
- because $T_2$ used the invalid value written by $T_1$, $T_2$ must also be rolled back
- $T_3$ must then be rolled back because it depended on $T_2$ and so on

This chain reaction of dependent rollbacks is called a cascading rollback.

#### Rigorous two-phase locking

Rigorous two-phase locking is a stricter version of the two-phase locking protocol.

With rigorous two-phase locking:

- all locks (both shared and exclusive) are held until the transaction either commits or aborts
- no locks are released during the execution of the transaction

This guarantees:

- serialisability
- no cascading rollbacks
- schedules are equivalent to a serial order based on commit order

Strict two-phase locking is a variant of this, where all exclusive locks are released at the end of the transaction, but shared  locks can be released earlier. 

### Deadlocks

Deadlocks are a problem that can occur when using locking-based concurrency control.

A deadlock occurs when two or more transactions are each waiting for a lock held by another, forming a cycle of waiting. As a result, they may wait indefinitely.

For example:

- $T_1$ gets the exclusive lock on $x$
- $T_2$ gets the exclusive lock on $y$
- $T_1$ requests an exclusive lock on $y$
- $T_2$ requests an exclusive lock on $x$
- neither request can be granted, so both transactions must wait
- each transaction is waiting for the other to release a lock, forming a circular wait (deadlock)

#### Handling deadlocks

One common method for handling deadlocks is timeouts:

- a transaction requests a lock and is only allowed to wait for a system-defined maximum time
- if the lock is not granted within this time:
  - the DBMS assumes there is a deadlock
  - the transaction request times out
  - the transaction rolls back and restarts

Another method is deadlock detection:

- the DBMS constructs a wait-for graph $G=(N,E)$ where:
  - each node $T_i \in N$ represents a transaction
  - a directed edge $T_i \to T_j$ exists if $T_i$ is waiting for a lock currently held by $T_j$
- a deadlock exists if and only if the wait-for graph contains a directed cycle
- when a deadlock is detected:
  - the DBMS selects one (or more) transactions as a victim
  - the victim transaction is aborted (rolled back) to break the cycle
