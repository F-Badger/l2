2. For a P2P file-sharing application, do you agree with the statement, "There is no notion of client and server sides of a communication session"? Why or why not?

No. There is a notion of client and server sides of a communication session, the specifics if which depend on the nature of the particular communication. If a device is sharing a file to another device, that device acts as a server. That same device can act as a client when it is downloading a file from another device, in which case the device it is downloading the file from now acts as the server. 

*No. At the level of socket communication, one side must first initiate communication with the other side (which must already be listening). In a P2P
file-sharing application, in the context of any given file transfer, the peer that is receiving the file is typically the client and the peer that is sending the file is typically the server.*

3. What information is used by a process running on one host to identify a process running on another host?

The IP address and socket number. 

*The IP address of the destination host and the port number to which the socket of the destination process is connected.*

4. Suppose you wanted to do a transaction from a remote client to a server as fast as possible. Would you use UDP or TCP? Why?

UDP. It has only been specified that the transaction needs to be done as fast as possible, not as reliably as possible. UDP is faster than TCP, since it does not establish a connection beforehand. 

*You would use UDP. With UDP, the transaction can be completed in one roundtrip time (RTT) - the client sends the transaction request into a UDP
socket, and the server sends the reply back to the client's UDP socket. With TCP, a minimum of two RTTs are needed - one to set-up the TCP connection,
and another for the client to send the request, and for the server to send back the reply.*

5. What is meant by a handshaking protocol?

A handshaking protocol is a protocol where two (or more) devices acknowledge or agree upon something. 

*A protocol uses handshaking if the two communicating entities first exchange control packets before sending data to each other. SMTP (for
example) uses handshaking at the application layer whereas HTTP does not.*

The server program must be executed before the client program to ensure the server has an open socket waiting for a connection request from the client. In UDP, there is no connection that needs to be established, so the client program can be executed before the server program. 

*For the TCP application, as soon as the client is executed, it attempts to initiate a TCP connection with the server. If the TCP server is not running at this point in time, then the client will fail to make a connection and will not be able to continue or send any data. In a similar UDP application, the client does not perform handshaking to establish a connection; when it sends its data, the call will succeed even if the server is not running. It would then continue to wait for a response from the server; if the server is running by the time the client sends its request (i.e. after waiting for user input), it would hopefully (i.e. no guarantee) get a response from the server.*
