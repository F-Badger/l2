# Networks

A computer network is a group of devices that are connected to one another in order to exchange information or share resources.

There are three main categories of network hardware:
- end devices (end systems, end stations, hosts)
  - end devices form the interface between users and the underlying communication network, e.g. computers, sensors, servers, etc. These are are the source or destination of communication
- intermediary devices
  - intermediary devices interconnect end devices, e.g. routers, switches, firewalls, wireless access points, etc. They manage, direct, or filter traffic as it moves across the network
- transmission media
  - transmission media provides the channels over which the message travels from source to destination, e.g copper cables, optic fibres, radio
frequency, etc. Data appears as signals in transmission media.

## Network protocols

A network protocol defines the rules for the format and order of messages sent and received among network entities, and actions taken on message transmission and receipt, enabling devices to communicate with each other.

The core requirements a protocol should handle are:
- sender and receiver identification: each message should show who it was sent by and who it is intended for
- common language and structure: devices must agree on the meaning of the data and the structure of the message
- speed and timing of delivery: how fast data can be sent, when devices are allowed to send and how long to wait for a reply
- confirmation/acknowledgment: whether the receiver should confirm receipt of the message, and what happens if no response is received

More specifically, a network protocol should define:
- message timing
   - when a device can send a message
   - how long it waits for a response
   - what happens if a response times out
- message delivery options
   - who receives the message, e.g. unicast (one receiver), multicast (selected group of receivers), broadcast (everyone on network receives)
- message size
   - minimum and maximum message size
   - how large messages are broken into smaller pieces
- message encoding
   - how data will be converted into a form suitable for transmission (encoded) and how the receiver can reverse the process (decode)
- message formatting and encapsulation
   - formatting defines the structure of the message
   - encapsulation defines how a message should be placed inside another as it moves down the protocol stack

## Circuit switching

Circuit switching is a communication method where a dedicated, fixed path (circuit) is established between two devices before any data is sent, and that path remains reserved for the entire duration of the communication. It is commonly used in traditional telephone networks. 

The devices will be able to use the full bandwidth of the channel for the entire duration of the communication, so the connection is reliable and consistent. However, this wastes resources when no data is being sent and does not scale well for data networks where there are a large number of users and traffic is sent in bursts rather than continuously. 

## Packet switching

Packet switching is a communication method where there is no dedicated path between the sender and receiver. Transmission links are shared among multiple users, and data is divided into small units called packets before transmission. Different packets may take different paths to reach the same destination to improve efficiency, reliability and speed (e.g. avoiding a congested or failed link).  

This uses transmission links more efficiently, as multiple users can share the same network simultaneously and scales well for large data networks. 

## OSI reference model

The OSI model is a conceptual framework that describes how data is transmitted across a network. It divides network communication into seven layers, each with specific responsibilities.
-  upper layers are more concerned with user services and applications
-  lower layers are more concerned with hardware and the actual transmission of data

| Layer                  | Main Function                                                                  |
| ---------------------- | ------------------------------------------------------------------------------ |
| Application Layer      | Provides network services to end-users and applications                       |
| Presentation Layer     | Translates, encrypts, or compresses data for applications                      |
| Session Layer          | Manages sessions or connections between applications                           |
| Transport Layer        | Provides end-to-end communication, reliability, and flow control               |
| Network Layer          | Handles routing and addressing of packets between networks                     |
| Data-Link Layer        | Provides reliable node-to-node data transfer, error detection, and framing      |
| Physical Layer         | Deals with the actual hardware transmission of raw bits over a medium          |

There are many advantages of using a layered model:
- changes in one layer don't affect other layers
- provides a common language to describe networking functions and capabilities
- assists protocol design, as protocols can be designed independently for each layer but work together as a protocol suite

## Network structure

A network can be divided into two main parts:
- **network edge**
  - the part of the network that interfaces directly with users
  - where end devices/hosts (clients or servers) connect to the network
    - clients are user devices like desktops, laptops, smartphones, etc.
    - servers provide services (e.g. web, email, storage) and are often located in data centers
- **network core**
  - the high-speed backbone that interconnects the network edges
  - consists of a mesh of interconnected routers, where packets are routed from one router to another across the network

Data is broken into packets at network edges. The network core uses packet switching so multiple edges can share the same core simultaneously. 

## Routing and forwarding

Routing is the process by which routers determine the best path for a packet to reach its destination.

Forwarding is the process by which network devices move a packet from an incoming interface to an outgoing interface toward its next hop (the next router the packet is to be moved to).

A routing algorithm is used to select the best paths for packets to reach the destination(s). There are many different ways to determine the best paths, e.g. delays, error rates, energy consumption, etc.

Switches use a switching table to forward frames:
- when a switch receives a frame, it looks at the source MAC address and  destination MAC address
  - it records which port that source MAC address came from in its switching table if it is not already there
- if the destination MAC address exists in the table, the switch forwards the frame only to the port associated with that MAC address
  - this is called unicast, because the frame is sent to a single, specific device
- if the destination MAC address is not in the table, the switch forwards the frame to all ports except the one it came from
  - this is called broadcast, used to locate the device and update the table

Store-and-forward is a technique in which an intermediary device receives and stores an entire data unit, performs error checks to prevent corrupted frames from propagating ( traveling through the transmission medium toward the next node or destination), and then forwards it to the next hop or destination.

Nodal delay is the total time a packet spends at a single network node (like a router or switch) before being forwarded to the next node. It is the sum of four main components: 
- **processing delay** - the time spent on checking bit errors and on determining the output link
- **queuing delay** - the time the packet spends waiting in a buffer before transmission
  - packets will need to queue to wait to be transmitted on the link when the arrival rate of packets exceeds the link's transmission rate
  - if the queue (buffer/memory) fills up, packets can be dropped (lost)
    - lost packets may be retransmitted by the previous node, by the source (end devices), or not at all
- **transmission delay** - the time required to push all the bits of a packet onto the communication link
  - it is calculated as $\frac{L}{R}$, where $L$ is the packet length in bits and $R$ is the link transmission rate in bits per second
- **propagation delay** - the time taken for the signal to travel across the physical link medium after the bits have been transmitted
  - it is calculated as $\frac{d}{s}$, where $d$ is the distance between the two nodes and $s$ is the propagation speed of the signal in the medium

For an end-to-end path, the total packet delay is the sum of the nodal delay across all nodes and links that the packet travels across. 

## Client-server architecture

### Client
Clients are devices requesting information or services. They:
- may be intermittently connected
- may have dynamic IP addresses
- do not communicate directly with other clients

### Server 
Servers are devices responding to client requests by providing data or services. They:
- are typically always-on
- use fixed (static) IP addresses
- are often hosted in data centres for scalability

## Peer-to-peer architecture
There is no a dedicated server in a peer-to-peer network.
- every connected device (a peer) can function as both a server and a client
- the roles of client and server are set on a per request basis
- new peers bring new service capacity, as well as new service demands
- peers are intermittently connected and change IP addresses

## Processes communicating
Processes are programs running within a host.
- within same host, two processes communicate using inter-process communication (defined by the OS).
- processes on different hosts communicate by exchanging messages

A socket is a software mechanism that allows a process to:
- send messages into the network
- receive messages from the network

A process sends messages through its socket to another process's socket, and receives messages from its socket that were sent by another process's socket.

Sockets act as the interface between the application layer and transport layer in the network stack.

## Transport layer protocols and services

The two protocols at the transport layer are:
- Transmission control protocol (TCP)
- User datagram protocol (UDP)

The common services offered by the two protocols are:
- segmentation
  - Sender side: breaks application messages into segments, passes to network layer
  - Receiver side: reassembles segments into messages, passes to application layer
- a way for application processes on different hosts to communicate as if they were directly connected
  - each application or process is assigned a port number to distinguish it from others on the same host
  - port numbers allow the transport layer to deliver the segment to the correct application process on the host

### TCP
There are services offered by TCP that make it differ from UDP:
- Connection-oriented
  - establishes a connection before data transmission using a three-way handshake
  - connection termination also involves a handshake on both sides to gracefully close the connection
- Reliable transport
  - ensures data is delivered correctly and in order
  - Features include:
    - reordering segments at the receiver
    - retransmission of lost segments
    - acknowledgments and timers to detect loss

- Flow control
  - uses window sizes to control the maximum number of bytes that can be sent before receiving an acknowledgment
  - prevents the sender from overwhelming the receiver

- Full-duplex connection
  - both ends of the connection can send and receive data simultaneously
 
### UDP
Accordingly, there are services offered by UDP that differ from TCP:
- Connectionless
  - sends data directly without establishing a connection or performing handshakes between sender and receiver

- Unreliable data transfer
  - does not guarantee delivery or ordering of datagrams
  - no retransmission for lost or corrupted data

- Minimal additional services
  - lacks features provided by TCP, such as flow control, congestion control, and built-in security
 
These are all negatives compared to TCP, however UDP does have the advantages that it is faster (lower latency) and has a smaller packet header.

## Multiplexing and demultiplexing

### Multiplexing (at sender)
A host may have multiple applications/processes sending data at the same time. Multiplexing lets the transport layer combine data from multiple processes and send it over the network using IP.
- each segment or datagram gets a transport layer header containing the source and destination port numbers
- these port numbers act as tags that identify which process the data came from and where it should go at the receiver

### Demultiplexing (at receiver) 
Demultiplexing uses the port numbers in the transport header to deliver each segment/datagram to the correct application/process.
