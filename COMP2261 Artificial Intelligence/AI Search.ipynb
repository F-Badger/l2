# Search problems

## Definition

Every (search) problem has six essential components:

1. Initial state  
   The state in which the problem-solving agent starts.

2. Actions and Transition Function   
   A description of the possible actions available to the agent, via a transition (or successor) function  $\phi$ which, for each state $x$ in the state space and each action $\alpha$, returns a (possibly empty) set of states reachable from $x$ using $\alpha$.

   - The state space is implicit in the description of the transition function.
   - The pair $(y, \alpha)$ is a successor of $x$ if  
     $$
     y \in \phi(x, \alpha)
     $$

3. Goal Test   
   A function that determines whether a given state is a goal state.

4. Step-Cost Function  
   A non-negative step-cost function  
   $$
   \sigma(x, \alpha, y)
   $$  
   which gives the cost of transitioning from state $x$ to state $y$ via action $\alpha$ (if such a transition is possible).

   
5. A solution to a problem   
   A sequence of actions (or equivalently, a path of states) that leads from the initial state to a goal state.

6. Optimal solution    
   A solution with the lowest (or highest) path cost among all possible solutions.
   - The path cost of a path is defined as the sum of its step costs:
  $$
  \text{Path cost} = \sum \sigma(x, \alpha, y)

## Search tree

A search tree is a tree data structure used by to represent the exploration of possible states in a search problem:
- Initially, the search tree consists of a single root node
  - The state associated with the root node is the initial state.
- The search tree is built incrementally by repeatedly expanding unexpanded leaf nodes.
  - Expanding a leaf node means generating all its successor states using the transition function $\phi(x, \alpha)$ and adding a child node for each successor
- If the same state $y$ can be reached from a state $x$ by different actions $\alpha$ and $\beta$, then:
  - Separate child nodes are created for each action, even though they represent the same state, because the search tree records paths, not just states.
- Each edge in the search tree corresponds to an action, and each node stores:
  - Its associated state
  - The action that led to it from its parent
- The set of all unexpanded leaf nodes is called the fringe.
- The choice of which fringe node to expand next is determined by the search strategy.
- A goal node is a node whose associated state satisfies the goal test (it is a goal state)
- When a goal node is found, the path from the root to that node represents a solution.

Note that the search tree is different from the state space graph, as the same state may appear multiple times in the search tree via different paths.
